```python
"""
The edges selected by the algorithm have two properties:

No two selected edges share a vertex → they form a matching.
Every unselected edge touches a selected edge → no more edges can be added.

Every time the algorithm picks an edge, it "claims" both of its endpoints and deletes every edge
that conflicts with that choice. Because of this, the chosen edges cannot overlap, and every
discarded edge is discarded only because it conflicts with a chosen edge. That is exactly what makes the set a maximal matching.
"""
def approx_vertex_cover(graph):
    edges = set()
    
    for u in graph:
        for v in graph[u]:
            edges.add(tuple(sorted((u, v))))

    cover = set()

    while edges:
        # Pick an arbitrary edge
        u, v = edges.pop()

        # Add both endpoints
        cover.add(u)
        cover.add(v)

        # Remove every remaining edge incident on u or v
        edges = {
            (a, b)
            for (a, b) in edges
            if u not in (a, b) and v not in (a, b)
        }

    return cover
```

```python
graph = {
    'A': {'B', 'C'},
    'B': {'A', 'C', 'D'},
    'C': {'A', 'B'},
    'D': {'B'}
}

cover = approx_vertex_cover(graph)
print(cover)
```

```python
"""
The term **incident** means that an edge is directly connected to a vertex.
In other words, a vertex is incident to every edge that has that vertex as one of its endpoints.
For example, the edge (A,B) is incident on both A and B, but it is not incident on C or D.

Consider the complete graph K₄:

      A
    / | \
   B--+--C
    \ | /
      D

The edges are:

(A,B), (A,C), (A,D), (B,C), (B,D), (C,D)

Suppose the approximation algorithm chooses the edge (A,B) first.

Step 1:
Add both endpoints to the vertex cover.

Cover = {A, B}

Step 2:
Remove every edge incident on A or B. This removes all edges that contain either A or B as an endpoint:

Remove:
(A,B)
(A,C)
(A,D)
(B,C)
(B,D)

The only remaining edge is:

(C,D)

Notice that (C,D) is not incident on A or B because neither endpoint is A or B.

Step 3:
The algorithm must choose the remaining edge (C,D).

Add both endpoints:

Cover = {A, B, C, D}

All edges have now been removed, so the algorithm terminates.

Final vertex cover = {A, B, C, D}

Why isn't {A, B} enough?

Although A and B cover every edge touching A or B, they do not cover the edge (C,D).
A vertex cover requires that every edge has at least one endpoint in the cover, and neither C nor D is in {A, B}.
Therefore, (C,D) remains uncovered.

The optimal vertex cover for K₄ has size 3. For example, {A, B, C} covers every edge because
every edge has at least one endpoint in the set. The only vertex not chosen is D, and every edge
involving D is covered by A, B, or C. Thus, the approximation algorithm always returns 4 vertices,
while the optimal solution requires only 3, making the algorithm's solution suboptimal for K₄.
"""
```

```python
def tree_vertex_cover(graph, root):
    cover = set()

    def dfs(node, parent):
        has_uncovered_child = False

        for neighbor in graph[node]:
            if neighbor != parent:
                child_covered = dfs(neighbor, node)

                # If a child was not covered, we need to choose this node
                if not child_covered:
                    has_uncovered_child = True

        if has_uncovered_child:
            cover.add(node)
            return True

        return node in cover

    dfs(root, None)

    return cover
```

```python
tree = {
    'A': {'B', 'C'},
    'B': {'A', 'D', 'E'},
    'C': {'A', 'F'},
    'D': {'B'},
    'E': {'B'},
    'F': {'C'}
}

result = tree_vertex_cover(tree, 'A')

print(result)
```

```python
"""
## Optimal Vertex Cover for a Tree Using a Greedy Algorithm

For a general graph, finding the minimum vertex cover is NP-complete, but for trees we can find
an optimal solution efficiently in linear time. The key observation is that if we have a leaf node in a tree,
its only edge must be covered by either the leaf or its parent. Choosing the parent is always
at least as good as choosing the leaf because the parent can cover the leaf's edge and possibly many other edges connected to it.
Therefore, there is always an optimal solution that contains the parent of every selected leaf.

The algorithm works by rooting the tree and processing it from the leaves upward. During a postorder traversal, we check each node.
If a node is a child of a vertex that is not already covered, we add the parent to the vertex cover.
By doing this repeatedly, every edge becomes covered while avoiding unnecessary vertices.

Example:

Consider the tree:

        A
       / \
      B   C
     / \   \
    D   E   F

Initially:

Cover = {}

The leaves are D, E, and F.

For leaf D:
The edge (B,D) must be covered. Instead of choosing D, choose its parent B because B also covers the edge (B,E) and possibly other edges.

Cover = {B}

For leaf F:
The edge (C,F) must be covered. Choose its parent C.

Cover = {B,C}

Now every edge is covered:
- (B,D) is covered by B.
- (B,E) is covered by B.
- (A,B) is covered by B.
- (C,F) is covered by C.
- (A,C) is covered by C.

The final vertex cover is:

{B,C}

This is optimal because any vertex cover must choose at least one endpoint of every edge.
The greedy algorithm always replaces a leaf with its parent, which can only cover the same edge or more edges.
Therefore, the algorithm never produces a larger solution than necessary.

The algorithm runs in O(V) time because every vertex and edge is processed only a constant number of times.
"""
```
