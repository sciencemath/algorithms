````python
"""
The edges selected by the algorithm have two properties:

No two selected edges share a vertex → they form a matching.
Every unselected edge touches a selected edge → no more edges can be added.

Every time the algorithm picks an edge, it "claims" both of its endpoints and deletes every edge
that conflicts with that choice. Because of this, the chosen edges cannot overlap, and every
discarded edge is discarded only because it conflicts with a chosen edge. That is exactly what makes the set a maximal matching.
"""
def approx_vertex_cover(graph):
    edges = set()
    
    for u in graph:
        for v in graph[u]:
            edges.add(tuple(sorted((u, v))))

    cover = set()

    while edges:
        # Pick an arbitrary edge
        u, v = edges.pop()

        # Add both endpoints
        cover.add(u)
        cover.add(v)

        # Remove every remaining edge incident on u or v
        edges = {
            (a, b)
            for (a, b) in edges
            if u not in (a, b) and v not in (a, b)
        }

    return cover
````

````python
graph = {
    'A': {'B', 'C'},
    'B': {'A', 'C', 'D'},
    'C': {'A', 'B'},
    'D': {'B'}
}

cover = approx_vertex_cover(graph)
print(cover)
````

````python
"""
Example Walkthrough

Consider the graph with the following edges:

(A,B), (A,C), (B,C), (B,D)

Initially:

Cover = {}
Remaining Edges = {(A,B), (A,C), (B,C), (B,D)}

Step 1:
Choose an arbitrary edge. Suppose the algorithm selects (A,B).

Add both endpoints to the cover.

Cover = {A, B}

Step 2:
Since both A and B are now in the cover, every edge incident on A or B is already covered. Remove all such edges.

Remove:
(A,B)
(A,C)
(B,C)
(B,D)

Remaining Edges = {}

Step 3:
There are no edges left, so the algorithm terminates.

Final Vertex Cover = {A, B}

Explanation:

The returned set {A, B} is a valid vertex cover because every original edge has at least one endpoint in the set:
- (A,B) is covered by A or B.
- (A,C) is covered by A.
- (B,C) is covered by B.
- (B,D) is covered by B.

Although this may not always be the minimum vertex cover, the approximation algorithm guarantees that the size of the
returned cover is at most twice the size of the optimal (minimum) vertex cover.
"""
````

````python
"""
The term **incident** means that an edge is directly connected to a vertex.
In other words, a vertex is incident to every edge that has that vertex as one of its endpoints.
For example, the edge (A,B) is incident on both A and B, but it is not incident on C or D.

Consider the complete graph K₄:

      A
    / | \
   B--+--C
    \ | /
      D

The edges are:

(A,B), (A,C), (A,D), (B,C), (B,D), (C,D)

Suppose the approximation algorithm chooses the edge (A,B) first.

Step 1:
Add both endpoints to the vertex cover.

Cover = {A, B}

Step 2:
Remove every edge incident on A or B. This removes all edges that contain either A or B as an endpoint:

Remove:
(A,B)
(A,C)
(A,D)
(B,C)
(B,D)

The only remaining edge is:

(C,D)

Notice that (C,D) is not incident on A or B because neither endpoint is A or B.

Step 3:
The algorithm must choose the remaining edge (C,D).

Add both endpoints:

Cover = {A, B, C, D}

All edges have now been removed, so the algorithm terminates.

Final vertex cover = {A, B, C, D}

Why isn't {A, B} enough?

Although A and B cover every edge touching A or B, they do not cover the edge (C,D).
A vertex cover requires that every edge has at least one endpoint in the cover, and neither C nor D is in {A, B}.
Therefore, (C,D) remains uncovered.

The optimal vertex cover for K₄ has size 3. For example, {A, B, C} covers every edge because
every edge has at least one endpoint in the set. The only vertex not chosen is D, and every edge
involving D is covered by A, B, or C. Thus, the approximation algorithm always returns 4 vertices,
while the optimal solution requires only 3, making the algorithm's solution suboptimal for K₄.
"""
````

````python
def tree_vertex_cover(graph, root):
    cover = set()

    def dfs(node, parent):
        has_uncovered_child = False

        for neighbor in graph[node]:
            if neighbor != parent:
                child_covered = dfs(neighbor, node)

                # If a child was not covered, we need to choose this node
                if not child_covered:
                    has_uncovered_child = True

        if has_uncovered_child:
            cover.add(node)
            return True

        return node in cover

    dfs(root, None)

    return cover
````

````python
tree = {
    'A': {'B', 'C'},
    'B': {'A', 'D', 'E'},
    'C': {'A', 'F'},
    'D': {'B'},
    'E': {'B'},
    'F': {'C'}
}

result = tree_vertex_cover(tree, 'A')

print(result)
````

````python
"""
## Optimal Vertex Cover for a Tree Using a Greedy Algorithm

For a general graph, finding the minimum vertex cover is NP-complete, but for trees we can find
an optimal solution efficiently in linear time. The key observation is that if we have a leaf node in a tree,
its only edge must be covered by either the leaf or its parent. Choosing the parent is always
at least as good as choosing the leaf because the parent can cover the leaf's edge and possibly many other edges connected to it.
Therefore, there is always an optimal solution that contains the parent of every selected leaf.

The algorithm works by rooting the tree and processing it from the leaves upward. During a postorder traversal, we check each node.
If a node is a child of a vertex that is not already covered, we add the parent to the vertex cover.
By doing this repeatedly, every edge becomes covered while avoiding unnecessary vertices.

Example:

Consider the tree:

        A
       / \
      B   C
     / \   \
    D   E   F

Initially:

Cover = {}

The leaves are D, E, and F.

For leaf D:
The edge (B,D) must be covered. Instead of choosing D, choose its parent B because B also covers the edge (B,E) and possibly other edges.

Cover = {B}

For leaf F:
The edge (C,F) must be covered. Choose its parent C.

Cover = {B,C}

Now every edge is covered:
- (B,D) is covered by B.
- (B,E) is covered by B.
- (A,B) is covered by B.
- (C,F) is covered by C.
- (A,C) is covered by C.

The final vertex cover is:

{B,C}

This is optimal because any vertex cover must choose at least one endpoint of every edge.
The greedy algorithm always replaces a leaf with its parent, which can only cover the same edge or more edges.
Therefore, the algorithm never produces a larger solution than necessary.

The algorithm runs in O(V) time because every vertex and edge is processed only a constant number of times.
"""
````

````python
"""
Clique: everyone is connected to everyone.
Independent set: nobody is connected to anyone.
Vertex cover: a set that touches every edge.

Clique = "everyone knows everyone."
Independent set = "nobody knows anyone."
Vertex cover = "pick a few people that touch every friendship."
"""
````

````python
G = {
    'A': {'B':2, 'C':3, 'D':6, 'E':7},
    'B': {'A':2, 'C':1, 'D':4, 'E':3},
    'C': {'A':3, 'B':1, 'D':5, 'E':2},
    'D': {'A':6, 'B':4, 'C':5, 'E':2},
    'E': {'A':7, 'B':3, 'C':2, 'D':2},
}
````

````python
import heapq

def mst_prim(G, r):
    parent = {v: None for v in G}
    key = {v: float("inf") for v in G}
    visited = set()
    key[r] = 0
    pq = [(0, r)]

    while pq:
        _, u = heapq.heappop(pq)

        if u in visited:
            continue

        visited.add(u)

        for v, weight in G[u].items():
            if v not in visited and weight < key[v]:
                key[v] = weight
                parent[v] = u
                heapq.heappush(pq, (weight, v))

    return parent
````

````python
from collections import defaultdict

def build_tree(parent):
    tree = defaultdict(list)

    for child, p in parent.items():
        if p is not None:
            tree[p].append(child)

    return tree
````

````python
def preorder(tree, root):
    order = []

    def dfs(u):
        order.append(u)
        for v in tree[u]:
            dfs(v)

    dfs(root)
    return order
````

````python
def approx_tsp_tour(G):
    root = next(iter(G))          # choose any vertex
    parent = mst_prim(G, root)    # compute MST
    tree = build_tree(parent)     # convert to child lists
    tour = preorder(tree, root)   # preorder traversal
    tour.append(root)             # return to the start
    return tour
````

````python
tour = approx_tsp_tour(G)
print(tour)
````

# Why Does This Work?

Suppose we insert a new vertex u.

Its closest vertex already on the tour is v.

Originally the tour contained

```
... → v → w → ...
```

After insertion

```
... → v → u → w → ...
```

The increase in cost is

```
c(v,u)+c(u,w)-c(v,w)
```

Because the triangle inequality says

```
c(u,w)
≤
c(u,v)+c(v,w)
```

we have

```
Increase

=
c(v,u)+c(u,w)-c(v,w)

≤

c(v,u)+c(u,v)

=

2c(u,v)
```

Thus **every insertion increases the tour length by at most**

```
2 × (distance from u to its nearest vertex already on the tour).
```

---

# Where Do Those Distances Come From?

Every inserted vertex connects to exactly one previous vertex.

Those chosen edges form a tree.

Example:

```
        A
       / \
      B   D
      |
      C
      |
      E
```

The total weight of these edges is exactly

```
Σ c(u,parent(u))
```

Since this is a spanning tree,

```
Tree Cost

≥

MST
```

or equivalently,

```
MST
≤
Tree Cost.
```

Also,

```
MST
≤
OPT
```

because deleting one edge from an optimal TSP tour produces a spanning tree.

Therefore

```
Tree Cost
≤
OPT.
```

---

# Total Cost

Each insertion contributes at most

```
2c(u,parent(u))
```

Therefore

```
Tour Cost

≤

2 Σ c(u,parent(u))

=

2(Tree Cost)

≤

2(MST)

≤

2OPT.
```

Hence

```
Tour Cost ≤ 2OPT.
```

---

# Final Proof

Let u be the next vertex inserted into the current cycle, and let v be its closest vertex already on the cycle. Suppose u is inserted between v and another vertex w.

The increase in the tour length is

```
c(v,u)+c(u,w)-c(v,w).
```

By the triangle inequality,

```
c(u,w) ≤ c(u,v)+c(v,w),
```

so

```
c(v,u)+c(u,w)-c(v,w)
≤
2c(u,v).
```

Thus each insertion increases the tour by at most twice the distance from u to its nearest previously inserted vertex.

The nearest-neighbor edges chosen during the insertions form a spanning tree. Their total weight is at most the weight of an optimal traveling-salesperson tour because the MST has minimum weight among all spanning trees, and the MST is no heavier than an optimal TSP tour.

Therefore,

```
Tour Cost
≤
2(Tree Cost)
≤
2(MST)
≤
2OPT.
```

Hence the closest-point insertion heuristic always produces a tour whose total cost is at most twice the cost of an optimal tour.

````python
def greedy_set_cover(X, F):
    uncovered = set(X)
    cover = []
    step = 1

    while uncovered:
        print(f"\nStep {step}")
        print("Uncovered:", uncovered)

        best_set = max(F, key=lambda s: len(s & uncovered))
        covered = best_set & uncovered

        if not covered:
            print("No set can cover any remaining elements.")
            break

        print("Choose:", best_set)
        print("Newly covered:", covered)

        cover.append(best_set)
        uncovered -= best_set

        print("Remaining:", uncovered)

        step += 1

    return cover

# Example
X = {1, 2, 3, 4, 5, 6, 7}

F = [
    {1, 2, 3},
    {2, 4, 5},
    {3, 4, 6},
    {5, 6, 7},
    {7}
]

result = greedy_set_cover(X, F)

print("\nSelected sets:")
for s in result:
    print(s)
````

# Summary

The Greedy Set Cover algorithm follows these steps:

1. Start with every element uncovered.
2. Find the set that covers the largest number of uncovered elements.
3. Add that set to the solution.
4. Remove those covered elements from the uncovered set.
5. Repeat until no uncovered elements remain.

The algorithm does **not always produce the minimum possible number of sets**, but it is guaranteed to produce a solution that is close to optimal and runs efficiently.

# Theorem

The decision version of the SET-COVER problem is NP-complete.

## Proof

We prove this in two parts.

--------------------------------------------------
Part 1: SET-COVER is in NP
--------------------------------------------------

An instance of the decision version of SET-COVER consists of

- A universe X.
- A family of subsets F = {S1, S2, ..., Sm}, where each Si is a subset of X.
- An integer k.

The question is:

"Is there a collection of at most k subsets whose union equals X?"

Suppose someone gives us a certificate consisting of the chosen subsets.

To verify the certificate:

1. Check that no more than k subsets were selected.
2. Compute the union of the selected subsets.
3. Verify that this union equals X.

Each of these operations can be performed in polynomial time.

Therefore, SET-COVER belongs to NP.

--------------------------------------------------
Part 2: Reduction from VERTEX-COVER
--------------------------------------------------

Suppose we are given an instance of the VERTEX-COVER problem.

Input:

G = (V, E)

and an integer k.

The question is:

"Does G have a vertex cover of size at most k?"

We construct an equivalent SET-COVER instance.

--------------------------------------------------
Construction
--------------------------------------------------

Universe

Let the universe consist of all edges of the graph.

X = E

Every edge is now an element that must be covered.

--------------------------------------------------
Subsets
--------------------------------------------------

For every vertex v in V, create one subset Sv.

Sv contains every edge incident to v.

Formally,

Sv = { e in E : e is incident to v }

The family of subsets is

F = {Sv : v in V}

Keep the same value of k.

--------------------------------------------------
Why does this work?
--------------------------------------------------

Each subset represents choosing one vertex.

Each edge appears in exactly the subsets corresponding to its endpoints.

Therefore,

choosing a subset

is equivalent to

choosing its corresponding vertex.

An edge is covered by the chosen subsets

if and only if

at least one of its endpoints was chosen.

But this is exactly the definition of a vertex cover.

--------------------------------------------------
Correctness
--------------------------------------------------

(⇒)

Suppose G has a vertex cover C with

|C| ≤ k.

For every vertex in C, choose the corresponding subset.

Since every edge has at least one endpoint in C,

every edge belongs to at least one chosen subset.

Thus all elements of X are covered.

Therefore the constructed SET-COVER instance has a cover of size at most k.

--------------------------------------------------

(⇐)

Suppose the SET-COVER instance has a cover consisting of at most k subsets.

For every chosen subset, select its corresponding vertex.

Since every edge is covered by one of the chosen subsets,

every edge has at least one selected endpoint.

Therefore these vertices form a vertex cover of size at most k.

--------------------------------------------------
Polynomial Time
--------------------------------------------------

The reduction creates

- one universe element for every edge, and
- one subset for every vertex.

Constructing each subset requires examining the incident edges.

Thus the reduction runs in polynomial time.

--------------------------------------------------
Conclusion
--------------------------------------------------

We have shown

1. SET-COVER belongs to NP.
2. VERTEX-COVER ≤p SET-COVER.

Since VERTEX-COVER is NP-complete,

SET-COVER is also NP-complete.

````python
"""
The improvement comes from avoiding recomputing every intersection after each greedy choice. Instead:

Compute the initial counts once.
Maintain a reverse mapping from elements to the sets that contain them.
When an element is covered, decrement the counts only for the sets containing that element.
Use a max-priority queue to efficiently retrieve the set with the largest remaining count.
"""
def greedy_set_cover(X, F):
    # count[S] = number of uncovered elements currently in S
    count = {}
    # reverse[e] = list of sets containing element e
    reverse = {}
    # Build reverse lookup and initial counts
    for S in F:
        count[id(S)] = len(S)

        for e in S:
            if e not in reverse:
                reverse[e] = []
            reverse[e].append(S)

    cover = []
    uncovered = set(X)

    while uncovered:
        # Find set with largest count
        best = None
        best_count = -1

        for S in F:
            if count[id(S)] > best_count:
                best = S
                best_count = count[id(S)]

        if best_count == 0:
            break

        cover.append(best)

        # Remove every newly covered element
        for e in list(best):

            if e not in uncovered:
                continue

            uncovered.remove(e)

            # Update only the affected sets
            for T in reverse[e]:
                count[id(T)] -= 1

    return cover
````

````python
"""
Here we compute x̄, an optimal solution to the linear-programming relaxation

linprog here is a "black box" we dont care how it computes we just need 0.5 rounding to either be 1 or 0

you treat the LP solver as an oracle: it returns the optimal fractional solution, and your job is to perform
the simple rounding step of selecting every vertex whose LP value is at least 0.5.
This is the part that proves the 2-approximation guarantee.
"""
from scipy.optimize import linprog

def approx_min_weight_vc(vertices, edges, weights):
    """
    vertices: list of vertices
    edges: list of (u, v)
    weights: dictionary mapping vertex -> weight

    Returns:
        Approximate minimum-weight vertex cover
    """

    n = len(vertices)
    index = {v: i for i, v in enumerate(vertices)}

    # Objective function
    c = [weights[v] for v in vertices]

    A = []
    b = []

    # Every edge must satisfy
    # x_u + x_v >= 1
    #
    # linprog expects
    #
    # Ax <= b
    #
    # so rewrite as
    #
    # -x_u - x_v <= -1

    for u, v in edges:
        row = [0] * n
        row[index[u]] = -1
        row[index[v]] = -1
        A.append(row)
        b.append(-1)

    bounds = [(0, 1) for _ in vertices]

    result = linprog(c, A_ub=A, b_ub=b, bounds=bounds)

    x = result.x

    cover = []

    for i, value in enumerate(x):
        if value >= 0.5:
            cover.append(vertices[i])

    return cover, x
````

````python
"""
Randomized 2-approximation for MAX-CNF-SAT.

Every variable is assigned True or False uniformly at random.
A clause fails only when every one of its literals is false.
A one-literal clause is satisfied with probability 1/2, which is the worst case.
Clauses with more literals are even more likely to be satisfied.
Therefore, on average, the algorithm satisfies at least half of all clauses.
Since the optimal assignment cannot satisfy more than all the clauses, this gives a 2-approximation.
"""
import random

def random_max_cnf_assignment(variables):
    assignment = {}

    for var in variables:
        assignment[var] = random.choice([True, False])

    return assignment
````

````python
"""
Randomized 2-approximation algorithm for MAX-CUT.

Returns the two sides of the cut.
"""
import random

def random_max_cut(vertices):
    S = set()
    T = set()

    for v in vertices:
        if random.random() < 0.5:
            S.add(v)
        else:
            T.add(v)

    return S, T
````

````python
"""
Merge two sorted lists into one sorted list
without duplicates.
"""
def merge_lists(L1, L2):
    return sorted(set(L1 + L2))


def exact_subset_sum(S, t):
    L = [[0]]

    for x in S:
        added = [value + x for value in L[-1]]
        merged = merge_lists(L[-1], added)
        merged = [value for value in merged if value <= t]
        L.append(merged)

    return max(L[-1]), L
````

````python
"""
We don't need every value. If two values are almost the same, keep only one.
this makes the algorithm polynomial instead of exponential (as opposed to the subset_sum)

Suppose
delta = 0.20

This means
20%

If two numbers differ by less than 20%, we throw one away.
"""
def trim(L, delta):
    if not L:
        return []

    trimmed = [L[0]]
    last = L[0]

    for value in L[1:]:
        if value > last * (1 + delta):
            trimmed.append(value)
            last = value

    return trimmed
````

````python
def approx_subset_sum(S, t, epsilon):
    """
    Approximate Subset Sum (CLRS)

    S       : list of positive integers
    t       : target sum
    epsilon : approximation parameter

    Returns the largest subset sum <= t.
    """

    n = len(S)

    # L current list of subset sums
    L = [0]

    for x in S:
        # Add x to every existing sum
        new_values = [y + x for y in L]

        # Merge old and new sums
        L = merge_lists(L, new_values)

        # Trim nearly identical values
        L = trim(L, epsilon / (2 * n))

        # Remove sums larger than the target
        L = [y for y in L if y <= t]

    return max(L)
````

````python
"""
| Exact Algorithm            | Approximation Algorithm                                   |
| -------------------------- | --------------------------------------------------------- |
| Generate all possible sums | Generate all possible sums                                |
| Keep every sum             | Remove sums that are almost duplicates (`trim`)           |
| Continue                   | Continue                                                  |
| Return best answer         | Return an answer that's guaranteed to be close to optimal |

The intuition behind e / (2n)

One subtle point is why the algorithm trims using e/(2n) instead of just e.

Each call to trim() introduces a small multiplicative error. Since the algorithm trims once for each of the n numbers,
those small errors accumulate. By making each trimming step use only a fraction of the total allowed error,
CLRS proves that after all n iterations, the final answer is still guaranteed to be within a factor of (1−ε)
of the optimal solution. In other words:

trim() at each step introduces a tiny approximation error.
Using e/(2n) keeps each individual error very small.
After all iterations, the total error stays below the user-specified tolerance e.

This is the key idea that turns the exact exponential-time subset sum algorithm into a
Fully Polynomial-Time Approximation Scheme (FPTAS): by carefully controlling the error introduced at each step,
the algorithm runs in polynomial time while still returning a provably near-optimal solution.

Every time you call

trim(...)

you lose a tiny bit of accuracy.

Trim #1  → lose a tiny bit
Trim #2  → lose a tiny bit
Trim #3  → lose a tiny bit
...
Trim #n  → lose a tiny bit
If you lost 20% every time, after many trims your answer could be far from the optimum.

If
epsilon = 0.20

and
n = 5

then
delta = epsilon / (2 * n)
      = 0.20 / 10
      = 0.02

Each trim only throws away numbers within about 2% of each other instead of 20%.

Think of it like a budget.

Total approximation error allowed

20%
│
├── iteration 1 uses about 2%
├── iteration 2 uses about 2%
├── iteration 3 uses about 2%
├── iteration 4 uses about 2%
└── iteration 5 uses about 2%

Each trimming step is more conservative than seems necessary, because CLRS wants a rigorous proof that after all n trims,
the overall approximation guarantee still holds.

So your observation was exactly right: the simple "budget" analogy isn't mathematically accurate.
The real reason for 2n comes from controlling the product of many small multiplicative errors,
and the proof uses exponential inequalities to ensure the final error stays within the desired bound.
"""
````

````python
"""
# Why does CLRS use δ = ε / (2n)?

At first, it seems like the algorithm should use

    δ = ε / n

because there are n calls to trim().

However, the approximation errors do NOT add together—they multiply.

------------------------------------------------------------
Step 1: Error from one trim
------------------------------------------------------------

Suppose a value y is removed during one call to trim().

The algorithm guarantees there is a remaining value z such that

    y ≤ z(1 + δ)

This means one trimming operation introduces at most a multiplicative error of

    1 + δ

------------------------------------------------------------
Step 2: Error after n trims
------------------------------------------------------------

Since trim() is called once for each element of the input, there are n trimming operations.

The total error is therefore

    (1 + δ)^n

Notice that the errors multiply together—they do NOT add.

------------------------------------------------------------
Step 3: Choose δ
------------------------------------------------------------

CLRS chooses

    δ = ε / (2n)

Substituting this into the previous equation gives

    (1 + ε/(2n))^n

------------------------------------------------------------
Step 4: Use a standard inequality
------------------------------------------------------------

From calculus, we know

    1 + x ≤ e^x

Therefore,

    (1 + ε/(2n))^n ≤ e^(ε/2)

------------------------------------------------------------
Step 5: Bound the exponential
------------------------------------------------------------

For

    0 < ε ≤ 1

another standard inequality states

    e^(ε/2) ≤ 1 + ε

Combining the two inequalities gives

    (1 + ε/(2n))^n ≤ 1 + ε

This proves that after all n trimming operations, the total approximation error is at most ε.

------------------------------------------------------------
Example: Why not use ε / n?
------------------------------------------------------------

Suppose

    ε = 0.20
    n = 5

If we use

    δ = ε / n = 0.04

then the accumulated error becomes

    (1.04)^5 = 1.2167

This corresponds to about a 21.7% error.

Since the user only allowed a 20% approximation error, this is too large.

------------------------------------------------------------

Now use the CLRS choice

    δ = ε / (2n)
      = 0.20 / 10
      = 0.02

The accumulated error becomes

    (1.02)^5 = 1.104

This corresponds to about a 10.4% error.

Since

    10.4% < 20%

the approximation guarantee is satisfied.

------------------------------------------------------------
Intuition
------------------------------------------------------------

Each call to trim() throws away values that are very close together, introducing a small approximation error.

If each trim were allowed to introduce a large error, those errors would compound over the n iterations.

To prevent this, CLRS chooses

    δ = ε / (2n)

so that every trimming step is very conservative.

The result is that, even after all n trimming operations, the final answer is guaranteed to be within the
user-specified approximation factor ε.

------------------------------------------------------------
The Main Takeaway
------------------------------------------------------------

Many students think:

    "There are n trims, so each trim should use ε/n."

But this ignores that approximation errors COMPOUND rather than simply adding together.

CLRS therefore uses the smaller value

    δ = ε / (2n)

which is a conservative choice that makes the mathematical proof work. The extra factor of 2 provides enough
safety margin to guarantee that, after all n trimming operations, the final subset sum is still within the requested approximation ratio.
"""
````

````python
"""
set implementation:
"""
def trim(sorted_list, delta):
    if not sorted_list:
        return []

    result = [sorted_list[0]]
    last = sorted_list[0]

    for value in sorted_list[1:]:
        if value > last * (1 + delta):
            result.append(value)
            last = value

    return result


def approx_subset_sum(S, t, epsilon):
    n = len(S)
    delta = epsilon / (2 * n)
    L = {0}

    for x in S:
        # Generate new subset sums
        new_sums = {s + x for s in L if s + x <= t}

        # Union removes duplicates
        L |= new_sums

        # Sort before trimming
        L = set(trim(sorted(L), delta))

    return max(L)
````

````python
"""
list implementation:
"""
def merge_lists(L1, L2):
    merged = []
    i = j = 0

    while i < len(L1) and j < len(L2):
        if L1[i][0] <= L2[j][0]:
            merged.append(L1[i])
            i += 1
        else:
            merged.append(L2[j])
            j += 1

    merged.extend(L1[i:])
    merged.extend(L2[j:])

    return merged


def trim(L, delta):
    if not L:
        return []

    trimmed = [L[0]]
    last_sum = L[0][0]

    for current_sum, subset in L[1:]:
        if current_sum > last_sum * (1 + delta):
            trimmed.append((current_sum, subset))
            last_sum = current_sum

    return trimmed


def approx_subset_sum(S, t, epsilon):
    n = len(S)
    delta = epsilon / (2 * n)

    # L starts with sum 0 and the empty subset
    L = [(0, [])]

    for x in S:

        # Create new sums by adding x
        new_list = []
        for current_sum, subset in L:
            new_list.append((current_sum + x, subset + [x]))

        # Merge old and new lists
        L = merge_lists(L, new_list)

        # Remove sums greater than t
        L = [item for item in L if item[0] <= t]

        # Trim nearby sums
        L = trim(L, delta)

    # Largest remaining sum
    best_sum, best_subset = L[-1]

    return best_sum, best_subset

S = [2, 4, 5]
t = 7
epsilon = 0.4

best_sum, subset = approx_subset_sum(S, t, epsilon)

print(best_sum)
print(subset)
````

````python
"""
In the reduction from SUBSET-SUM to BIN-PACKING:

B = target value from the SUBSET-SUM problem.
T = total sum of all numbers in the SUBSET-SUM input.

The two values have different meanings:

B tells us the exact sum we are trying to find.
T tells us the sum of every number in the entire input.

Example:

SUBSET-SUM instance:

Numbers:
2, 5, 8, 10

Target:
B = 13

First compute the total:

T = 2 + 5 + 8 + 10
T = 25

So:

B = 13 (the desired subset sum)
T = 25 (the sum of all numbers)

Now convert the SUBSET-SUM instance into a BIN-PACKING instance.

For each number a_i, create an object with size:

s_i = a_i / T

The object sizes are:

2/25
5/25
8/25
10/25

Then add one extra object with size:

(T - B) / T

Substitute the values:

(25 - 13) / 25

= 12/25

The final bin-packing objects are:

2/25
5/25
8/25
10/25
12/25

Why do we add an object of size (T-B)/T?

Suppose the SUBSET-SUM answer is YES.

The subset:

5 + 8 = 13

has size:

5/25 + 8/25

= 13/25

Now place these objects together with the extra object:

13/25 + 12/25

= 25/25

= 1

Therefore, this subset exactly fills one bin.

The remaining objects are:

2/25 + 10/25

= 12/25

which fit into the second bin.

The key idea:

B = the amount we want to find.

T-B = the amount left over after finding that subset.

The extra object has size:

(T-B)/T

because it perfectly fills the remaining space:

B/T + (T-B)/T = 1

So if a subset with sum B exists, the objects can be packed into exactly two bins.

Conversely, if the objects can be packed into two bins, the objects placed with the extra object must fill exactly:

1 - (T-B)/T

= B/T

Multiplying by T gives:

subset sum = B

Therefore, solving BIN-PACKING would allow us to solve SUBSET-SUM.

Since SUBSET-SUM is NP-complete, BIN-PACKING is NP-hard.
"""
````

````python
def first_fit(items):
    bins = []

    for item in items:

        placed = False

        for i in range(len(bins)):
            if bins[i] + item <= 1:
                bins[i] += item
                placed = True
                break

        if not placed:
            bins.append(item)

    return bins

"""
Bin 1:
0.6 + 0.4 = 1

Bin 2:
0.8 + 0.2 = 1

Bin 3:
0.3
"""
items = [0.6, 0.4, 0.8, 0.3, 0.2]

print(first_fit(items))
````
