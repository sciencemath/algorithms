# Proof that LONGEST-PATH-LENGTH is Polynomial-Time Solvable iff LONGEST-PATH ∈ P

## Proof

We prove both directions.

### (⇒) If LONGEST-PATH-LENGTH is solvable in polynomial time, then LONGEST-PATH ∈ P.

Assume there exists a polynomial-time algorithm `A(G, u, v)` that returns the length `L` of the longest simple path from `u` to `v`.

To solve an instance `<G, u, v, k>` of LONGEST-PATH:

1. Run `A(G, u, v)`.
2. Let the returned value be `L`.
3. If `L >= k`, accept; otherwise reject.

Since computing `L` takes polynomial time and the comparison `L >= k` takes constant time, the entire algorithm runs in polynomial time.

Therefore, **LONGEST-PATH ∈ P.**

---

### (⇐) If LONGEST-PATH ∈ P, then LONGEST-PATH-LENGTH is solvable in polynomial time.

Assume there exists a polynomial-time decision algorithm `B(G, u, v, k)` that returns **TRUE** if there exists a simple path from `u` to `v` containing at least `k` edges, and **FALSE** otherwise.

Let `n = |V|`. A simple path in an undirected graph can visit each vertex at most once, so its maximum possible length is `n - 1` edges.

Therefore, the answer must lie between `0` and `n - 1`.

Use binary search on this range.

Initialize:

```
low = 0
high = n - 1
```

While `low < high`:

1. Compute

```
mid = (low + high + 1) // 2
```

2. Run the decision algorithm `B(G, u, v, mid)`.

* If it returns **TRUE**, then there exists a path of length at least `mid`, so set

```
low = mid
```

* Otherwise, set

```
high = mid - 1
```

When the binary search finishes, `low == high`, and this value is exactly the length of the longest simple path.

---

## Running Time

The search space contains at most `n` possible path lengths.

Binary search requires `O(log n)` iterations.

Each iteration calls the polynomial-time decision algorithm once.

Therefore, the total running time is

```
O(log n) × poly(n) = poly(n).
```

Hence LONGEST-PATH-LENGTH is solvable in polynomial time.

---

## Conclusion

We have shown both directions:

* If LONGEST-PATH-LENGTH can be solved in polynomial time, then LONGEST-PATH is in `P`.
* If LONGEST-PATH is in `P`, then LONGEST-PATH-LENGTH can be solved in polynomial time.

Therefore,

**LONGEST-PATH-LENGTH is solvable in polynomial time if and only if LONGEST-PATH ∈ P.**

---

## Intuition

The optimization problem computes the exact longest path length.

The decision problem only answers whether a path of length at least `k` exists.

* **Optimization → Decision:** Compute the exact answer and compare it with `k`.
* **Decision → Optimization:** Ask a sequence of yes/no questions using binary search to determine the largest feasible path length.

Thus, the optimization and decision problems are polynomial-time equivalent.

# Closure of the Class P

Assume that

- $L_1 \in P$
- $L_2 \in P$

This means there exist polynomial-time algorithms $A_1$ and $A_2$ that decide membership in $L_1$ and $L_2$, respectively.

We show that $P$ is closed under union, intersection, concatenation, complement, and Kleene star.

---

## 1. Closure Under Union

We want to show

$$
L_1 \cup L_2 \in P.
$$

### Algorithm

Given an input string $x$:

1. Run $A_1(x)$.
2. Run $A_2(x)$.
3. Accept if either algorithm accepts; otherwise reject.

### Running Time

Suppose

$$
A_1 = O(n^a), \qquad
A_2 = O(n^b).
$$

The total running time is

$$
O(n^a) + O(n^b)
=
O(n^{\max(a,b)}),
$$

which is polynomial.

Therefore,

$$
L_1 \cup L_2 \in P.
$$

---

## 2. Closure Under Intersection

We want to show

$$
L_1 \cap L_2 \in P.
$$

### Algorithm

Given $x$:

1. Run $A_1(x)$.
2. Run $A_2(x)$.
3. Accept only if both algorithms accept.

### Running Time

Again,

$$
O(n^a) + O(n^b)
=
O(n^{\max(a,b)}),
$$

which is polynomial.

Therefore,

$$
L_1 \cap L_2 \in P.
$$

---

## 3. Closure Under Complement

We want to show

$$
\overline{L_1} \in P.
$$

### Algorithm

Given $x$:

1. Run $A_1(x)$.
2. If $A_1$ accepts, reject.
3. If $A_1$ rejects, accept.

Only the final answer is flipped.

### Running Time

The running time is exactly that of $A_1$, which is polynomial.

Therefore,

$$
\overline{L_1} \in P.
$$

---

## 4. Closure Under Concatenation

We want to show

$$
L_1L_2 \in P.
$$

A string belongs to $L_1L_2$ if it can be written as

$$
x = yz,
$$

where

$$
y \in L_1
\quad\text{and}\quad
z \in L_2.
$$

### Algorithm

Given a string $x$ of length $n$:

For every possible split position

$$
i = 0,1,\ldots,n,
$$

split the string into

- Prefix: $x[1:i]$
- Suffix: $x[i+1:n]$

Run

- $A_1$ on the prefix.
- $A_2$ on the suffix.

If both accept for any split, accept.

Otherwise reject.

### Running Time

There are

$$
n+1
$$

possible splits.

Each split performs two polynomial-time tests.

Thus the total running time is

$$
O(n)\times\text{poly}(n)
=
\text{poly}(n).
$$

Therefore,

$$
L_1L_2 \in P.
$$

---

## 5. Closure Under Kleene Star

We want to show

$$
L_1^* \in P.
$$

A string belongs to $L_1^*$ if it can be partitioned into zero or more substrings, each of which belongs to $L_1$.

### Algorithm

Let

$$
dp[i]
$$

mean:

> The first $i$ characters of the string can be written as a concatenation of strings from $L_1$.

Initialize

$$
dp[0] = \text{TRUE}.
$$

For each position $i$:

- Check every earlier position $j < i$.
- If

$$
dp[j] = \text{TRUE}
$$

and the substring

$$
x[j+1:i]
$$

belongs to $L_1$ (determined by running $A_1$),

then set

$$
dp[i] = \text{TRUE}.
$$

Finally, accept if

$$
dp[n] = \text{TRUE}.
$$

### Running Time

There are

$$
O(n^2)
$$

possible substrings.

Each substring is tested in polynomial time.

Thus the total running time is

$$
O(n^2)\times\text{poly}(n)
=
\text{poly}(n).
$$

Therefore,

$$
L_1^* \in P.
$$

---

# Conclusion

We have shown that

- $L_1 \cup L_2 \in P$
- $L_1 \cap L_2 \in P$
- $\overline{L_1} \in P$
- $L_1L_2 \in P$
- $L_1^* \in P$

Each construction requires only polynomially many operations and calls to polynomial-time algorithms.

Therefore, the complexity class **$P$ is closed under union, intersection, complement, concatenation, and Kleene star.**
