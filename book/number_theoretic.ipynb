```python
"""
Efficient binary-to-decimal conversion using divide-and-conquer.

Idea:
A binary number x can be split into two halves:
    x = x_high * 2^m + x_low

We recursively convert each half, then combine them using
multiplication and addition.

This yields the recurrence:
    T(B) = 2T(B/2) + O(M(B))

which solves to:
    O(M(B) log B)

Convert binary string to integer using divide-and-conquer
"""

def binary_to_decimal(bin_str):
    n = len(bin_str)

    if n == 1:
        return int(bin_str)

    mid = n // 2
    high = bin_str[:mid]
    low = bin_str[mid:]

    high_val = binary_to_decimal(high)
    low_val = binary_to_decimal(low)

    # Combine: high * 2^(len(low)) + low
    return high_val * (1 << len(low)) + low_val


examples = [
    "11010110",   # 214
    "1011",       # 11
    "100000",     # 32
    "11111111"    # 255
]

for b in examples:
    print(b, "->", binary_to_decimal(b))
```

```python
def euclid(a, b):
    if b == 0:
        return a
    else:
        return euclid(b, a % b)
```

```python
def euclid_extended(a, b):
    if b == 0:
        return (a, 1, 0)
    else:
        d, x1, y1 = euclid_extended(b, a % b)
        x = y1
        y = x1 - (a // b) * y1
        return (d, x, y)
```

```python
def euclid_iterative(a, b):
    while b != 0:
        a, b = b, a % b
    return a
```

```python
def lcm(a, b):
    return abs(a * b) // gcd(a, b)

def lcm_n(nums):
    result = 1
    for x in nums:
        result = lcm(result, x)
    return result
```

```python
def modular_exponentiation(a, b, n):
    if b == 0:
        return 1 % n
    elif b % 2 == 0:
        d = modular_exponentiation(a, b // 2, n)  # b is even
        return (d * d) % n
    else:
        d = modular_exponentiation(a, b - 1, n)  # b is odd
        return (a * d) % n
```

```python
def modular_exponentiation_iterative(a, b, n):
    result = 1
    a = a % n

    while b > 0:
        if b % 2 == 1:
            result = (result * a) % n
        a = (a * a) % n
        b //= 2

    return result
```

```python
"""
Using

else:
    d = modular_exponentiation(a, b // 2, n)
    return (d * d * a) % n

instead,

You reduce b to b//2, i.e. shift right by 1 bit.
So if b has B bits, recursion depth is exactly B.
No "linear peeling" like b−1, which can cause extra calls.
"""
def modular_exponentiation(a, b, n):
    if b == 0:
        return 1 % n

    d = modular_exponentiation(a, b // 2, n)
    d = (d * d) % n

    if b % 2 == 1:
        d = (d * a) % n

    return d
```

```python
"""
Knowing d lets you compute φ(n), then p+q, and finally factor n via a quadratic equation.
Every step is efficient (polynomial time in the bit-length of n).

PROOF: If e = 3 and the adversary knows d (0 < d < φ(n)), then they can factor n in polynomial time.

Given:
n = pq where p, q are primes
φ(n) = (p − 1)(q − 1)
e = 3
d satisfies: 3d ≡ 1 (mod φ(n))

STEP 1: Convert modular equation into an integer equation
3d ≡ 1 (mod φ(n))
⇒ 3d − 1 is a multiple of φ(n)
⇒ 3d − 1 = kφ(n) for some integer k

WHY: This removes modular arithmetic and gives a usable exact equality.

STEP 2: Substitute the structure of φ(n)
φ(n) = (p − 1)(q − 1)
      = pq − p − q + 1
      = n − (p + q) + 1

So:
3d − 1 = k(n − (p + q) + 1)

WHY: We rewrite φ(n) in terms of n, p+q so we can eventually recover p and q.

STEP 3: Isolate (p + q)
3d − 1 = k(n + 1 − (p + q))

Rearrange:
(p + q) = n + 1 − (3d − 1)/k

WHY: If we can determine φ(n), we immediately get p+q, and then we can factor n.

STEP 4: Key observation — d is small and forces structure
Since 0 < d < φ(n), the value (3d − 1) is not arbitrary; it is a structured multiple of φ(n).
Thus we compute:
g = gcd(3d − 1, n)

WHY: Any nontrivial relation between 3d − 1 and n must reveal a factor of n, because:
- 3d − 1 = k(p − 1)(q − 1)
- (p − 1)(q − 1) shares hidden structure with p and q

STEP 5: Extract a factor
Compute:
g = gcd(3d − 1, n)

Then:
- if g = p or g = q → we have factored n
- otherwise, refine using:
  gcd(3d − 1, p − 1) or gcd(3d − 1, q − 1) logic (standard RSA reduction step)

WHY: gcd reveals hidden shared prime structure between constructed multiple and n.

STEP 6: Recover full factorization
Once one prime is found:
p = g
q = n / p

WHY: n = pq, so division gives the other factor immediately.

CONCLUSION:
Knowing d lets an attacker compute a structured multiple of φ(n).
That structure leaks nontrivial gcd information with n, allowing recovery of p or q in polynomial time.
Therefore, RSA is insecure if the secret exponent d is revealed.
"""
```

```python
"""
Fermat primality test with base 2.

If n is prime, the procedure always returns 'prime'.

If n is composite:
    - Usually returns 'composite'.
    - May incorrectly return 'prime' if n is a base-2 pseudoprime.

Base-2 pseudoprimes are rare. Examples:
341, 561, 645, 1105, ...
"""

def pseudoprime(n):
    if n < 2:
        return 'composite'

    if modular_exponentiation(2, n - 1, n) != 1:
        return 'composite'

    return 'prime'
```

```python
"""
True: a is a witness that n is composite.
False: a is not a witness (n could still be composite, need more bases).

1. Factor n-1 = 2^t * u
2. Compute x0 = a^u mod n
3. Repeatedly square
   x1 = x0² mod n
   x2 = x1² mod n
   ...
   xt = xt-1² mod n

4. During squaring:
      if we find a nontrivial square root of 1
      -> COMPOSITE

5. At the end:
      if a^(n-1) mod n != 1
      -> COMPOSITE

6. Otherwise:
      this base did not prove compositeness
"""
def witness(a, n):
    # n - 1 = 2^t * u where u is odd
    u = n - 1
    t = 0

    while u % 2 == 0:
        t += 1
        u //= 2

    x = pow(a, u, n)

    for _ in range(t):
        y = (x * x) % n

        if y == 1 and x != 1 and x != n - 1:
            return True

        x = y

    return x != 1
```

```python
def miller_rabin(n, s=20):
    if n < 2:
        return False
    if n in (2, 3):
        return True
    if n % 2 == 0:
        return False

    for _ in range(s):
        a = random.randint(2, n - 2)
        if witness(a, n):
            return False

    return True
```
