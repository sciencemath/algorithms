```python
"""
Efficient binary-to-decimal conversion using divide-and-conquer.

Idea:
A binary number x can be split into two halves:
    x = x_high * 2^m + x_low

We recursively convert each half, then combine them using
multiplication and addition.

This yields the recurrence:
    T(B) = 2T(B/2) + O(M(B))

which solves to:
    O(M(B) log B)

Convert binary string to integer using divide-and-conquer
"""

def binary_to_decimal(bin_str):
    n = len(bin_str)

    if n == 1:
        return int(bin_str)

    mid = n // 2
    high = bin_str[:mid]
    low = bin_str[mid:]

    high_val = binary_to_decimal(high)
    low_val = binary_to_decimal(low)

    # Combine: high * 2^(len(low)) + low
    return high_val * (1 << len(low)) + low_val


examples = [
    "11010110",   # 214
    "1011",       # 11
    "100000",     # 32
    "11111111"    # 255
]

for b in examples:
    print(b, "->", binary_to_decimal(b))
```

```python
def euclid(a, b):
    if b == 0:
        return a
    else:
        return euclid(b, a % b)
```

```python
def euclid_extended(a, b):
    if b == 0:
        return (a, 1, 0)
    else:
        d, x1, y1 = euclid_extended(b, a % b)
        x = y1
        y = x1 - (a // b) * y1
        return (d, x, y)
```

```python
def euclid_iterative(a, b):
    while b != 0:
        a, b = b, a % b
    return a
```

```python
def lcm(a, b):
    return abs(a * b) // gcd(a, b)

def lcm_n(nums):
    result = 1
    for x in nums:
        result = lcm(result, x)
    return result
```

```python
def modular_exponentiation(a, b, n):
    if b == 0:
        return 1 % n
    elif b % 2 == 0:
        d = modular_exponentiation(a, b // 2, n)  # b is even
        return (d * d) % n
    else:
        d = modular_exponentiation(a, b - 1, n)  # b is odd
        return (a * d) % n
```

```python
def modular_exponentiation_iterative(a, b, n):
    result = 1
    a = a % n

    while b > 0:
        if b % 2 == 1:
            result = (result * a) % n
        a = (a * a) % n
        b //= 2

    return result
```

```python
"""
Using

else:
    d = modular_exponentiation(a, b // 2, n)
    return (d * d * a) % n

instead,

You reduce b to b//2, i.e. shift right by 1 bit.
So if b has B bits, recursion depth is exactly B.
No "linear peeling" like b−1, which can cause extra calls.
"""
def modular_exponentiation(a, b, n):
    if b == 0:
        return 1 % n

    d = modular_exponentiation(a, b // 2, n)
    d = (d * d) % n

    if b % 2 == 1:
        d = (d * a) % n

    return d
```

```python
"""
Fermat primality test with base 2.

If n is prime, the procedure always returns 'prime'.

If n is composite:
    - Usually returns 'composite'.
    - May incorrectly return 'prime' if n is a base-2 pseudoprime.

Base-2 pseudoprimes are rare. Examples:
341, 561, 645, 1105, ...
"""

def pseudoprime(n):
    if n < 2:
        return 'composite'

    if modular_exponentiation(2, n - 1, n) != 1:
        return 'composite'

    return 'prime'
```

```python
"""
True: a is a witness that n is composite.
False: a is not a witness (n could still be composite, need more bases).

1. Factor n-1 = 2^t * u
2. Compute x0 = a^u mod n
3. Repeatedly square
   x1 = x0² mod n
   x2 = x1² mod n
   ...
   xt = xt-1² mod n

4. During squaring:
      if we find a nontrivial square root of 1
      -> COMPOSITE

5. At the end:
      if a^(n-1) mod n != 1
      -> COMPOSITE

6. Otherwise:
      this base did not prove compositeness
"""
def witness(a, n):
    # n - 1 = 2^t * u where u is odd
    u = n - 1
    t = 0

    while u % 2 == 0:
        t += 1
        u //= 2

    x = pow(a, u, n)

    for _ in range(t):
        y = (x * x) % n

        if y == 1 and x != 1 and x != n - 1:
            return True

        x = y

    return x != 1
```

```python
def miller_rabin(n, s=20):
    if n < 2:
        return False
    if n in (2, 3):
        return True
    if n % 2 == 0:
        return False

    for _ in range(s):
        a = random.randint(2, n - 2)
        if witness(a, n):
            return False

    return True
```

```python
"""
A prime power only has the trivial square roots 1 and −1.
But when a modulus has multiple distinct prime factors, the Chinese Remainder Theorem lets you
independently choose +1 on some factors and −1 on others. Mixing these choices creates new solutions
that square to 1 but are neither globally 1 nor globally −1. That's exactly why composite numbers with at least two
distinct prime factors always have nontrivial square roots of 1.
"""
```

```python
"""
Example 1: n = 15

We are given n = 15. Since

15 = 3 × 5,

15 is not prime and it is not a prime power (because it has two different prime factors).
The theorem says there should exist a nontrivial square root of 1 modulo 15.

The idea is to use the Chinese Remainder Theorem (CRT). Instead of trying random numbers, we intentionally construct one.

## Step 1: Choose congruences

We let

x ≡ 1 (mod 3)

and

x ≡ -1 (mod 5).

Since -1 mod 5 is the same as 4, this becomes

x ≡ 1 (mod 3)
x ≡ 4 (mod 5)

Why do we do this?

Because if x is congruent to ±1 modulo each prime factor, then squaring x will always produce 1 modulo each prime factor.

Specifically,

1² = 1

and

(-1)² = 1.

Therefore x² will equal 1 modulo both 3 and 5.

-------------------------------------------------------

## Step 2: Find a number satisfying both congruences

First list numbers that are congruent to 1 modulo 3:

1, 4, 7, 10, 13, ...

Now check which one is congruent to 4 modulo 5.

1 mod 5 = 1
4 mod 5 = 4 ✓

Therefore

x = 4.

-------------------------------------------------------

## Step 3: Verify x² ≡ 1 (mod 15)

Compute

4² = 16

Now reduce modulo 15.

16 mod 15 = 1

Therefore

4² ≡ 1 (mod 15).

-------------------------------------------------------

## Step 4: Check that it is nontrivial

The trivial square roots of 1 are

1 and -1.

Modulo 15,

-1 ≡ 14.

Our solution is

x = 4.

Since

4 ≠ 1
4 ≠ 14,

4 is a nontrivial square root of 1 modulo 15.

-------------------------------------------------------

Why did this work?

We purposely made x behave like

+1 modulo 3

and

-1 modulo 5.

Squaring removes the negative sign:

(+1)² = 1
(-1)² = 1

Since x² equals 1 modulo both prime factors, the Chinese Remainder Theorem tells us x² ≡ 1 modulo 15.
"""
```

```python
"""
# Example 2: n = 21

Now consider

21 = 3 × 7.

Again, 21 is not prime and it is not a prime power because it has two distinct prime factors.
The theorem says there should be a nontrivial square root of 1 modulo 21.
Again we construct one using the Chinese Remainder Theorem.

-------------------------------------------------------

## Step 1: Choose congruences

Let

x ≡ 1 (mod 3)

and

x ≡ -1 (mod 7).

Since

-1 mod 7 = 6,

this becomes

x ≡ 1 (mod 3)
x ≡ 6 (mod 7)

Why choose these?

Because squaring either 1 or -1 always gives

1.

So if x behaves like 1 modulo one prime factor and -1 modulo the other, then x² will equal 1 modulo both.

-------------------------------------------------------

## Step 2: Find x

List numbers congruent to 1 modulo 3:

1, 4, 7, 10, 13, 16, 19, ...

Now reduce each modulo 7.

1 mod 7 = 1
4 mod 7 = 4
7 mod 7 = 0
10 mod 7 = 3
13 mod 7 = 6 ✓

Therefore

x = 13.

-------------------------------------------------------

## Step 3: Verify x² ≡ 1 (mod 21)

Compute

13² = 169.

Now divide by 21.

169 = 8 × 21 + 1

Therefore

169 mod 21 = 1

so

13² ≡ 1 (mod 21).

-------------------------------------------------------

## Step 4: Verify that it is nontrivial

The trivial square roots are

1 and -1.

Modulo 21,

-1 ≡ 20.

Our solution is

13.

Since

13 ≠ 1
13 ≠ 20,

13 is a nontrivial square root of 1 modulo 21.

-------------------------------------------------------

Why did this work?

We intentionally made x satisfy

x ≡ 1 (mod 3)

and

x ≡ -1 (mod 7).

Squaring gives

1² = 1
(-1)² = 1.

Since x² equals 1 modulo both prime factors, the Chinese Remainder Theorem guarantees

x² ≡ 1 (mod 21).

This is exactly how the proof works in general. Whenever n has at least two distinct prime factors,
we can choose x to be 1 modulo one prime power and -1 modulo the others. Squaring removes the sign difference,
producing a nontrivial square root of 1 modulo n.
"""
```

```python
"""
Think of λ(n) as the smallest exponent that works for every number relatively prime to n,
while ϕ(n) is a larger exponent that always works by Euler's theorem.

Since the least common multiple of the prime-power exponents must divide their product, we always have

λ(n) ∣ ϕ(n).

For a number to "fake" being prime in Fermat's test (a Carmichael number),
this exponent must also divide n−1. That forces a very rigid structure:

No repeated prime factors (square free), because repeated primes introduce an extra factor
of p into λ(n) that would also have to divide n−1, creating a contradiction.
At least three distinct primes, because with only two primes, the divisibility
conditions force the two primes to be equal, which is impossible.

This is why Carmichael numbers are relatively rare despite there being infinitely many of them.
"""
```

```python
"""
# Example: The Smallest Carmichael Number (561)

The smallest Carmichael number is

561 = 3 × 11 × 17.

Our goal is to verify that 561 satisfies the definition of a Carmichael number.

Recall that a composite number n is a Carmichael number if

λ(n) divides (n − 1),

where λ(n) is the least common multiple (LCM) of the Euler totients of each prime factor.

-------------------------------------------------------

## Step 1: Factor n

First, factor the number into its prime factors.

561 = 3 × 11 × 17

Why do we do this?

Both Euler's totient function φ(n) and the Carmichael function λ(n) are computed using the prime factorization of n. Everything starts with knowing the prime factors.

Notice that each prime appears only once, so 561 is square free.

-------------------------------------------------------

## Step 2: Compute φ(p) for each prime factor

For a prime p,

φ(p) = p − 1.

Therefore,

φ(3) = 2

φ(11) = 10

φ(17) = 16

Why do we do this?

Euler's theorem tells us that every number relatively prime to p satisfies

a^(φ(p)) ≡ 1 (mod p).

These exponents are the building blocks for λ(n).

-------------------------------------------------------

## Step 3: Compute φ(561)

Multiply the totients together.

φ(561)

= φ(3) × φ(11) × φ(17)

= 2 × 10 × 16

= 320

Why do we multiply?

Since the prime factors are all distinct, Euler's totient function is multiplicative. This gives the exponent used in Euler's theorem for modulo 561.

-------------------------------------------------------

## Step 4: Compute λ(561)

The Carmichael function is the least common multiple of the individual totients.

λ(561)

= lcm(2, 10, 16)

To compute the LCM, factor each number.

2 = 2

10 = 2 × 5

16 = 2⁴

Take the largest power of every prime.

Largest power of 2 is 2⁴.

Largest power of 5 is 5.

Therefore,

λ(561)

= 2⁴ × 5

= 16 × 5

= 80

Why do we use the LCM instead of multiplying?

Euler's theorem gives an exponent that always works, but it is not necessarily the smallest one.

The Carmichael function λ(n) is the smallest exponent that works for every number relatively prime to n.

Think of λ(n) as the "minimum universal exponent."

-------------------------------------------------------

## Step 5: Check the Carmichael condition

Compute

561 − 1 = 560

Now determine whether λ(561) divides 560.

560 ÷ 80 = 7

Since the quotient is an integer,

80 divides 560.

Therefore,

λ(561) | (561 − 1)

and 561 satisfies the definition of a Carmichael number.

-------------------------------------------------------

## Step 6: Verify the properties proved in the theorem

Property 1: Square free

561 = 3 × 11 × 17

No prime factor is repeated.

Therefore,

561 is square free.

Why is this important?

If a prime square divided n (for example, if n contained 3²), then λ(n) would contain an extra factor of 3.
Since λ(n) must divide n − 1, this would force 3 to divide both n and n − 1, which is impossible because
consecutive integers cannot both be divisible by the same prime.

-------------------------------------------------------

Property 2: Product of at least three primes

561 has exactly three distinct prime factors.

3, 11, and 17.

Why can't a Carmichael number be the product of only two primes?

Suppose

n = pq.

The Carmichael condition forces

p − 1 to divide q − 1

and

q − 1 to divide p − 1.

The only way both can happen is if

p − 1 = q − 1,

which implies

p = q.

But p and q were supposed to be different primes.

This contradiction shows that a Carmichael number cannot have only two prime factors.

-------------------------------------------------------

## Final Intuition

The Carmichael function λ(n) is the smallest exponent that makes

a^λ(n) ≡ 1 (mod n)

for every number relatively prime to n.

To be a Carmichael number, this exponent must also divide n − 1, making n behave like a prime in
Fermat's primality test even though it is composite.

This condition is extremely restrictive.

A Carmichael number cannot contain repeated prime factors because repeated factors
introduce extra prime powers into λ(n), making it impossible for λ(n) to divide n − 1.

It also cannot be the product of only two primes because the divisibility conditions force the two primes to be equal, which is impossible.

Therefore, every Carmichael number must be square free and must contain at least three distinct prime factors. The smallest example is

561 = 3 × 11 × 17.
"""
```

```python
"""
# Example: Using a Nontrivial Square Root of 1 to Factor n

Suppose

n = 15

and

x = 4.

First verify that x is a nontrivial square root of 1 modulo 15.

Compute

4² = 16

Since

16 mod 15 = 1,

we have

4² ≡ 1 (mod 15).

Also,

4 ≠ 1 (mod 15)

and

4 ≠ 14 (mod 15),

where 14 is equivalent to −1 modulo 15.

Therefore, x = 4 is a nontrivial square root of 1 modulo 15.

-------------------------------------------------------

## Step 1: Rewrite the congruence

Since

x² ≡ 1 (mod n),

subtract 1 from both sides.

x² − 1 ≡ 0 (mod n)

Now factor the left side.

x² − 1 = (x − 1)(x + 1)

Therefore,

(x − 1)(x + 1) ≡ 0 (mod n).

Why do we do this?

Factoring splits one equation into two smaller pieces. The prime factors of n must now be divided between (x − 1) and (x + 1). This allows the greatest common divisor (gcd) to recover those factors.

-------------------------------------------------------

## Step 2: Compute gcd(x − 1, n)

Compute

gcd(4 − 1, 15)

= gcd(3, 15)

= 3.

The gcd is neither 1 nor 15.

Therefore, 3 is a nontrivial factor of 15.

-------------------------------------------------------

## Step 3: Compute gcd(x + 1, n)

Compute

gcd(4 + 1, 15)

= gcd(5, 15)

= 5.

Again, the gcd is neither 1 nor 15.

Therefore, 5 is another nontrivial factor of 15.

-------------------------------------------------------

## Final Intuition

The equation

x² ≡ 1 (mod n)

means

(x − 1)(x + 1)

is divisible by n.

Since x is a nontrivial square root, neither

x − 1

nor

x + 1

is divisible by all of n.

Instead, each contains some of n's prime factors.

The gcd extracts those shared prime factors, allowing us to factor n.

In this example,

gcd(3,15) = 3

and

gcd(5,15) = 5,

which are exactly the two prime factors of 15.
"""
```

```python
"""
===============================================================================
MILLER-RABIN PRIMALITY TEST - COMPLETE INTUITION AND EXAMPLES
===============================================================================

GOAL
----
Given a very large integer n, determine whether it is prime.

Checking every divisor up to sqrt(n) works for small numbers but is far too
slow for cryptography where numbers may have 1024, 2048, or even 4096 bits.

Instead, Miller-Rabin is a PROBABILISTIC primality test.

Rather than proving a number is prime, it repeatedly asks:

    "Can I find evidence that this number is composite?"

If enough random tests fail to find evidence, then the number is
"probably prime."

===============================================================================
STEP 1 - FERMAT'S LITTLE THEOREM
===============================================================================

Everything begins with Fermat's Little Theorem.

If p is prime,

    a^(p-1) ≡ 1 (mod p)

for every integer a where gcd(a,p)=1.

Example:

p = 7
a = 3

3^6 = 729

729 mod 7 = 1

The theorem works perfectly for primes.

This suggests the following primality test:

If

    a^(n-1) != 1 (mod n)

then n MUST be composite.

Unfortunately...

Some composite numbers also satisfy

    a^(n-1) ≡ 1 (mod n)

for many values of a.

Example:

561 = 3 × 11 × 17

Yet

a^560 ≡ 1 (mod 561)

for every a relatively prime to 561.

These are called Carmichael Numbers.

They fool Fermat's test.

===============================================================================
STEP 2 - THE BIG IDEA OF MILLER-RABIN
===============================================================================

Instead of only checking the FINAL answer

    a^(n-1)

Miller-Rabin watches HOW we arrive there.

First write

    n - 1 = 2^t * u

where u is odd.

Example:

n = 221

221 - 1 = 220

Factor out powers of two.

220 = 2² × 55

Therefore

t = 2
u = 55

Instead of computing

a^220

we first compute

a^55

and repeatedly square.

The chain looks like

a^55
↓

a^110
↓

a^220

Notice each exponent doubles.

===============================================================================
STEP 3 - WHY DO WE KEEP SQUARING?
===============================================================================

Suppose n really is prime.

Eventually we know

a^(n-1) ≡ 1 (mod n)

Suppose our squaring sequence looks like

x0 = a^55

↓

x1 = x0²

↓

x2 = x1² = 1

Now ask:

"What could x1 have been?"

Since

x1² ≡ 1

x1 is a square root of 1.

Here is the important theorem:

If n is prime,

the ONLY square roots of 1 are

    1

and

   -1

There are no others.

This fact is exactly what Miller-Rabin exploits.

===============================================================================
STEP 4 - NONTRIVIAL SQUARE ROOTS OF 1
===============================================================================

Suppose we discover

37

↓

1

after squaring.

This means

37² ≡ 1 (mod n)

But

37 != 1

and

37 != -1

That is impossible if n were prime.

Therefore

n MUST be composite.

This strange number (37 in this example) is called a

NONTRIVIAL SQUARE ROOT OF 1.

Finding one immediately proves that n is composite.

===============================================================================
STEP 5 - WHAT IS A WITNESS?
===============================================================================

Think of a courtroom.

A witness gives evidence.

In Miller-Rabin,

a witness is simply a number a that proves n is composite.

Example:

Suppose

n = 221

Choose

a = 38

If during repeated squaring we encounter

197

↓

1

where

197 != 1

197 != 220 (-1 mod 221)

then 38 has exposed that 221 cannot be prime.

Therefore

38 is called a WITNESS.

A witness is just evidence that the number is composite.

===============================================================================
STEP 6 - WHAT IS A NONWITNESS?
===============================================================================

Suppose instead we choose

a = 21

and every Miller-Rabin test passes.

Does that prove 221 is prime?

NO.

It only means

21 failed to expose the composite.

That number is called a

NONWITNESS

(or sometimes a STRONG LIAR).

It lied by making the composite number look prime.

===============================================================================
VISUAL IDEA
===============================================================================

Composite Number

          👤

      /     |      \

   a=2    a=5    a=17

    ✔       ✘        ✔

 Witness  Nonwitness Witness

Some random bases expose the composite.

Some don't.

===============================================================================
STEP 7 - WHY RANDOM BASES WORK
===============================================================================

One of the greatest results about Miller-Rabin says:

If n is composite,

AT LEAST

3/4

of all possible bases are witnesses.

At most

1/4

are nonwitnesses.

That means every random test has at least

75%

chance of catching a composite.

Probability all random tests fail:

1 test

(1/4)

4 tests

(1/4)^4 = 1/256

10 tests

(1/4)^10

≈ 1 in 1,048,576

20 tests

(1/4)^20

≈ 9 × 10^-13

The error decreases exponentially.

===============================================================================
STEP 8 - COMPLETE EXAMPLE
===============================================================================

Test

n = 21

First compute

20 = 2² × 5

Therefore

t = 2

u = 5

Choose

a = 2

Compute

x0 = 2^5 mod 21

2^5 = 32

32 mod 21 = 11

Now square repeatedly.

x1 = 11² mod 21

121 mod 21 = 16

Next square.

x2 = 16² mod 21

256 mod 21 = 4

We expected eventually to reach

1

but instead reached

4

Therefore

2^20 != 1 mod 21

The test immediately concludes

21 is composite.

So

a = 2

is a witness.

===============================================================================
STEP 9 - WHY WRITE n-1 = 2^t * u ?
===============================================================================

This is probably the hardest conceptual step.

Every repeated squaring doubles an exponent.

Starting with

u

(which is odd)

lets us generate

u

↓

2u

↓

4u

↓

8u

↓

...

↓

2^t u = n-1

This exactly recreates

a^(n-1)

using repeated squaring.

By watching every intermediate value, Miller-Rabin can detect impossible
square roots of 1 BEFORE reaching the final exponent.

That is something Fermat's test never checks.

Think of it this way.

Fermat only checks the destination.

Miller-Rabin watches the entire journey.

===============================================================================
STEP 10 - PRIME NUMBER THEOREM
===============================================================================

CLRS states

Pr{random n is prime}

≈ 1 / ln(n)

Why?

The Prime Number Theorem says that near a large number n,

about one out of every

ln(n)

numbers is prime.

So

Probability(random integer is prime)

≈

1 / ln(n)

===============================================================================
STEP 11 - WHY DOES CLRS WRITE 1.443/B ?
===============================================================================

Suppose n has

B bits.

Then approximately

n ≈ 2^B

Take natural logarithms.

ln(n)

=

ln(2^B)

=

B ln(2)

Since

ln(2)

≈

0.693

we obtain

1 / ln(n)

=

1 / (B × 0.693)

≈

1.443 / B

This explains the equation in CLRS.

===============================================================================
EXAMPLE
===============================================================================

Suppose B = 1024 bits.

Probability(random 1024-bit integer is prime)

≈

1.443 / 1024

≈

0.00141

About

0.14%

or roughly

1 out of every 710

random odd 1024-bit numbers is prime.

This is why RSA key generation repeatedly generates random odd numbers
and runs Miller-Rabin until one passes.

===============================================================================
BIG PICTURE
===============================================================================

Think of Miller-Rabin as a detective.

1. Pick a random base a.

2. Compute

       a^u mod n

   where

       n-1 = 2^t * u.

3. Repeatedly square.

4. If you ever find an impossible square root of 1,
   then you have found a witness.

5. A witness proves n is composite.

6. If the test never finds evidence, then a is a nonwitness.

7. Repeat using different random bases.

Because at least 75% of bases are witnesses for every composite number,
the chance that every random test misses the composite becomes
astronomically small.

This is why Miller-Rabin is one of the most important and widely used
primality tests in cryptography.
===============================================================================
"""
```

```python
"""
===============================================================================
WHY gcd(x − 1, n) AND gcd(x + 1, n) FIND FACTORS
===============================================================================

Suppose we have found a nontrivial square root of 1 modulo n.

That means

    x² ≡ 1 (mod n)

but

    x ≠ 1 (mod n)

and

    x ≠ -1 (mod n).

The first thing we do is factor the left-hand side:

    x² − 1 = (x − 1)(x + 1)

Since

    x² ≡ 1 (mod n),

we know

    n | (x² − 1)

which means

    n | (x − 1)(x + 1).

This tells us something incredibly important:

The PRODUCT of (x − 1) and (x + 1) contains every prime factor of n.

However, neither factor alone is divisible by n, because x is NOT ±1 modulo n.

So the prime factors of n must be "split" between the two numbers.

Think of n as a bag of prime factors.

Example:

    n = 21 = 3 × 7

Suppose

    x = 13

Then

    x − 1 = 12 = 2² × 3

    x + 1 = 14 = 2 × 7

Notice what happened.

The prime factor 3 went into (x − 1).

The prime factor 7 went into (x + 1).

Together they contain all the prime factors of 21.

Visual picture:

              n = 21

            /        \

           3          7
           |          |
           |          |
       x − 1      x + 1
         12          14

Now compute the gcds.

gcd(12,21)

= 3

gcd(14,21)

= 7

The gcd "pulls out" whichever prime factors of n are hiding inside that number.

Think of gcd as a filter.

                gcd(x − 1, n)

                      │

                      ▼

      Pull every common factor out of x − 1 and n

Similarly,

                gcd(x + 1, n)

                      │

                      ▼

      Pull every common factor out of x + 1 and n

Since the prime factors of n are split between x − 1 and x + 1,
each gcd recovers part of the factorization.

That is why

    gcd(x − 1, n)

and

    gcd(x + 1, n)

are guaranteed to be nontrivial divisors.

===============================================================================
THE BIG IDEA
===============================================================================

Finding a nontrivial square root of 1 does TWO things at once.

1. It proves n cannot be prime.
2. It immediately exposes hidden factors of n using the Euclidean Algorithm.

This is why Miller-Rabin watches for nontrivial square roots of 1.
If one ever appears, the number is definitely composite, and computing

    gcd(x − 1, n)

or

    gcd(x + 1, n)

often reveals the actual factors of n almost instantly.

This is one of the central ideas behind modern integer factorization and
primality testing algorithms.
===============================================================================
"""
```

```python
def binary_gcd(a, b):
    if a == 0:
        return b
    if b == 0:
        return a

    # Both numbers even
    if a % 2 == 0 and b % 2 == 0:
        return 2 * binary_gcd(a // 2, b // 2)

    # a even, b odd
    elif a % 2 == 0:
        return binary_gcd(a // 2, b)

    # a odd, b even
    elif b % 2 == 0:
        return binary_gcd(a, b // 2)

    # Both odd
    else:
        if a >= b:
            return binary_gcd((a - b) // 2, b)
        else:
            return binary_gcd((b - a) // 2, a)
```

```python
"""
Stein's Algorithm (Binary GCD)

The Binary GCD algorithm computes the greatest common divisor without using division or the remainder operator.
Instead, it relies on three fast operations that computers perform efficiently: checking whether a number is even or odd,
dividing by 2 (halving), and subtraction.

Algorithm Rules

1. If a == 0, return b.
2. If b == 0, return a.
3. If both numbers are even:
       gcd(a,b) = 2 × gcd(a/2, b/2)
   Since both numbers share a factor of 2, remove it, compute the gcd of the smaller numbers, and multiply the final answer by 2.

4. If one number is even and the other is odd:
       gcd(a,b) = gcd(a/2,b)
       or
       gcd(a,b) = gcd(a,b/2)
   The odd number cannot contribute a factor of 2 to the gcd, so the factor of 2 in the even number can simply be removed.

5. If both numbers are odd:
       gcd(a,b) = gcd((a-b)/2, b)
   (assuming a ≥ b)
   Subtracting preserves the gcd because any divisor common to both numbers also divides their difference.
   Since the difference of two odd numbers is even, divide it by 2 before continuing.

Python uses the modulo operator (%) to test parity:

    n % 2 == 0    # n is even
    n % 2 == 1    # n is odd

The algorithm repeatedly halves numbers or subtracts and halves them, causing the numbers to shrink rapidly.
Because the values decrease exponentially, the algorithm performs only O(log a) recursive calls,
making it as efficient as Euclid's algorithm while avoiding expensive remainder computations.
"""
```

```python
"""
Analysis of Bit Operations in Euclid's Algorithm

In this analysis, we count bit operations instead of arithmetic operations.
Arithmetic operations such as division become more expensive as the numbers contain more bits,
so we measure the true computational cost based on the number of bits being processed.

Part (a): Cost of Long Division

Suppose we divide a by b, producing

    a = bq + r

where q is the quotient and r is the remainder.

The quotient q contains approximately (1 + lg q) bits. The long division algorithm determines one quotient bit
during each iteration, so it performs O(1 + lg q) iterations.

During each iteration, the algorithm compares, subtracts, and shifts numbers that contain at most (lg b) bits.
Each of these operations requires O(lg b) bit operations.

Therefore,

    Total Cost
        = (number of iterations) × (cost per iteration)
        = O((1 + lg q) · lg b)

Thus ordinary binary long division requires

    O((1 + lg q) lg b)

bit operations.

------------------------------------------------------------

Part (b): The Potential Function

Define

    μ(a,b) = (1 + lg a)(1 + lg b).

Euclid transforms

    (a,b)

into

    (b, a mod b).

The division cost from Part (a) is

    O((1 + lg q) lg b),

where q = floor(a/b).

Since

    lg q ≈ lg a − lg b,

the division cost is proportional to

    (lg a − lg b)(lg b).

Now examine the change in the potential function:

    μ(a,b) − μ(b,r)

where r = a mod b.

Expanding,

    μ(a,b) − μ(b,r)
    = (1 + lg b)(lg a − lg r).

Because r < b,

    lg r < lg b,

so

    lg a − lg r
    ≥ lg a − lg b.

Therefore the decrease in μ is at least a constant multiple of the work performed during one division.

Hence there exists a constant c > 0 such that

    Cost of one Euclid step
    ≤ c(μ(a,b) − μ(b,r)).

The potential function decreases enough to pay for each recursive call.

------------------------------------------------------------

Part (c): Total Running Time

Each recursive call decreases the potential μ.

Summing the costs over every recursive call produces a telescoping sum:

    (μ0 − μ1)
  + (μ1 − μ2)
  + (μ2 − μ3)
  + ...

All intermediate terms cancel, leaving only

    μ(initial) − μ(final).

Since μ(final) is nonnegative,

    Total Cost = O(μ(a,b)).

Finally, if both inputs contain B bits, then

    1 + lg a = O(B)
    1 + lg b = O(B),

so

    μ(a,b)
    = (1 + lg a)(1 + lg b)
    = O(B²).

Therefore Euclid's algorithm performs

    O(B²)

bit operations on two B-bit integers.

Key Ideas

• Long division computes one quotient bit per iteration.
• Each iteration manipulates O(lg b) bits.
• The potential function μ(a,b) always decreases.
• The decrease in μ is large enough to pay for each division.
• Summing these decreases forms a telescoping sum, giving an overall runtime of O(μ(a,b)).
• For B-bit inputs, this simplifies to O(B²) bit operations.
"""
```

```python
def fib(n, memo={}):
    if n in memo:
        return memo[n]

    if n <= 1:
        return n

    memo[n] = fib(n-1, memo) + fib(n-2, memo)
    return memo[n]
```

```python
"""
Define the matrix

        [0 1]
    A = [1 1].

A fundamental identity is

           [F(n−1)  F(n)  ]
    Aⁿ =   [F(n)    F(n+1)].

Therefore, computing Aⁿ immediately gives F(n) as the upper-right matrix entry.

Instead of multiplying A by itself n times, use exponentiation by squaring.

Exponentiation by squaring repeatedly squares the matrix:

    A
    A²
    A⁴
    A⁸
    A¹⁶
    ...

Each squaring doubles the exponent, so only O(log n) matrix multiplications are required.

Since each matrix multiplication is constant time under the original assumptions,

    Running Time = O(log n).
"""
def matrix_mult(A, B):
    return [
        [
            A[0][0]*B[0][0] + A[0][1]*B[1][0],
            A[0][0]*B[0][1] + A[0][1]*B[1][1]
        ],
        [
            A[1][0]*B[0][0] + A[1][1]*B[1][0],
            A[1][0]*B[0][1] + A[1][1]*B[1][1]
        ]
    ]


def matrix_power(M, n):
    if n == 0:
        return [[1,0],[0,1]]

    if n == 1:
        return M

    if n % 2 == 0:
        half = matrix_power(M, n//2)
        return matrix_mult(half, half)

    return matrix_mult(M, matrix_power(M, n-1))


def fib(n):
    if n == 0:
        return 0

    A = [[0,1],[1,1]]

    return matrix_power(A,n)[0][1]
```

```python
"""
A quadratic residue is simply a number that you can obtain by squaring another number modulo p.

Think of taking every nonzero number modulo p, squaring it, and writing down the answers.
Every value that appears is a quadratic residue. Every value that never appears is a quadratic nonresidue.

Because x² and (−x)² are always the same modulo p, the nonzero numbers naturally pair up as {x, −x}.
Each pair produces one distinct square, so exactly half of the nonzero numbers are quadratic residues.
"""
```

```python
"""
Quadratic Residues

Let p be an odd prime, and let Z*p = {1, 2, ..., p−1}.
A number a ∈ Z*p is called a quadratic residue modulo p if there exists some x such that

    x² ≡ a (mod p).

Otherwise, a is called a quadratic nonresidue.

------------------------------------------------------------
(a) Show there are exactly (p − 1)/2 quadratic residues.
------------------------------------------------------------

Consider the squaring function

    f(x) = x² (mod p)

for every x in Z*p.

Suppose

    x² ≡ y² (mod p).

Then

    x² − y² ≡ 0 (mod p)

which factors as

    (x − y)(x + y) ≡ 0 (mod p).

Since p is prime,

    x ≡ y (mod p)

or

    x ≡ −y (mod p).

Thus every quadratic residue has exactly two square roots:

    x and −x.

These are different because p is odd (the only number equal to its own negative modulo p is 0, which is not in Z*p).

Since there are p−1 nonzero elements and they are grouped into pairs

    {x, −x},

the number of distinct quadratic residues is

    (p − 1)/2.

Therefore,

    There are exactly (p − 1)/2 quadratic residues modulo p.

------------------------------------------------------------
(b) Euler's Criterion and the Legendre Symbol
------------------------------------------------------------

The Legendre symbol is defined by

            { 1   if a is a quadratic residue
(a/p) =
            { -1  otherwise

Euler's Criterion states that

    (a/p) ≡ a^((p−1)/2) (mod p).

Proof

Case 1: a is a quadratic residue.

Then

    a ≡ x² (mod p)

for some x.

Raise both sides to the (p−1)/2 power:

    a^((p−1)/2)
    ≡ (x²)^((p−1)/2)
    = x^(p−1).

By Fermat's Little Theorem,

    x^(p−1) ≡ 1 (mod p),

so

    a^((p−1)/2) ≡ 1 (mod p).

Case 2: a is a quadratic nonresidue.

Fermat's theorem gives

    a^(p−1) ≡ 1 (mod p).

Therefore

    (a^((p−1)/2))² ≡ 1 (mod p).

The only square roots of 1 modulo a prime are

    ±1.

Since a is not a quadratic residue, the value cannot be 1.

Hence

    a^((p−1)/2) ≡ −1 (mod p).

Therefore,

    (a/p) = a^((p−1)/2) (mod p).

------------------------------------------------------------
Algorithm to Test Whether a is a Quadratic Residue
------------------------------------------------------------

Compute

    x = a^((p−1)/2) mod p

using Modular Exponentiation.

If

    x == 1

then a is a quadratic residue.

Otherwise

    x == p−1

(which represents −1 modulo p),

so a is a quadratic nonresidue.

Python

def quadratic_residue(a, p):
    x = pow(a, (p - 1) // 2, p)
    return x == 1

Running Time

Modular exponentiation uses repeated squaring.

The exponent has O(log p) bits.

Therefore the algorithm performs

    O(log p)

modular multiplications.

------------------------------------------------------------
(c) Finding Square Roots when p = 4k + 3
------------------------------------------------------------

Suppose

    p = 4k + 3

and a is a quadratic residue.

Let

    x = a^(k+1) mod p.

Then

    x²
    = a^(2k+2).

Since

    p = 4k + 3,

we have

    2k + 2
    = (p + 1)/2.

Therefore

    x²
    = a^((p+1)/2)
    = a · a^((p−1)/2).

Because a is a quadratic residue,

Euler's Criterion gives

    a^((p−1)/2) ≡ 1 (mod p).

Hence

    x² ≡ a (mod p).

Thus

    a^((p+1)/4)

is a square root of a modulo p.

Algorithm

def modular_sqrt(a, p):
    return pow(a, (p + 1) // 4, p)

Running Time

Again, only one modular exponentiation is required.

Running time:

    O(log p)

modular multiplications.

------------------------------------------------------------
(d) Randomized Algorithm for Finding a Nonquadratic Residue
------------------------------------------------------------

Exactly half of the nonzero numbers modulo p are quadratic residues.

The other half are nonresidues.

Therefore a randomly chosen element has

    Probability = 1/2

of being a nonquadratic residue.

Algorithm

1. Randomly choose

       a ∈ {1,...,p−1}.

2. Compute

       a^((p−1)/2) mod p.

3. If the result is

       p−1 (which is −1 mod p),

   then return a.

4. Otherwise choose another random number and repeat.

Python

import random

def random_nonresidue(p):
    while True:
        a = random.randint(1, p - 1)
        if pow(a, (p - 1) // 2, p) == p - 1:
            return a

Running Time

Each trial succeeds with probability

    1/2.

Therefore the expected number of trials is

    2.

Each trial performs one modular exponentiation requiring

    O(log p)

operations.

Expected running time:

    O(log p)

------------------------------------------------------------
Key Ideas (Aha Moment)

• Squaring pairs every nonzero number with its negative, so exactly half of the nonzero elements are distinct quadratic residues.
• Euler's Criterion determines whether a number is a quadratic residue by computing a^((p−1)/2) mod p. The result is 1 for residues and −1 for nonresidues.
• When p = 4k + 3, a square root of a quadratic residue can be computed directly as a^((p+1)/4) mod p.
• Since exactly half of the elements are nonresidues, repeatedly choosing random numbers and testing them with Euler's Criterion finds a nonquadratic residue in only two expected trials, giving an expected running time of O(log p).
"""
```
