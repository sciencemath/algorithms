```python
"""
Optimal overall does not imply optimal at every step

OPT = cost of the best possible algorithm that knows everything in advance (offline).
ALG = cost of your online algorithm.

A 2-competitive algorithm means
ALG ≤ 2⋅OPT

for every possible input sequence.

No matter how badly the future surprises me,
I will never do more than twice as badly as someone who knew the future.

Example

Imagine you're driving.
GPS (offline algorithm) knows all future traffic.
You (online algorithm) only know traffic when you encounter it.

Suppose:
GPS finds a route taking 20 minutes.
You take 35 minutes.

Ratio:
35/20 = 1.75.

That's okay for a 2-competitive algorithm.
If one day:

GPS takes 20
You take 50

Ratio:
50/20 = 2.5.

Then you're not 2-competitive.

What is an offline algorithm?
Offline means:
You know the entire input before you start.

For scheduling:
You know:
every release time
every processing time
every future job

before making your first decision.

This is usually unrealistic but gives us a benchmark.

What is an online algorithm?
Online means:
You only know what has arrived so far.

When a job arrives:
you learn its release time
you learn its processing time

but you know nothing about future jobs.

This is the realistic setting.

Why compare against offline?
Because offline is the strongest possible opponent.
If you can stay close to someone who literally knows the future,
your algorithm is pretty good.

Competitive analysis asks:
How much does not knowing the future hurt?
"""
```

```python
# Frequency-count example
L = ["A", "B", "C", "D"]
freq = {x: 0 for x in L}

def reorder():
    global L
    # Sort by decreasing frequency
    L = sorted(L, key=lambda x: (-freq[x], L.index(x)))

def search(x):
    global L, freq
    freq[x] += 1
    reorder()

    print(f"searched {x}")
    print("frequencies:", freq)
    print("list:", L)
    print()

search("D")

search("C")
search("C")

search("B")
search("B")
search("B")
```

```python
L = ["A", "B", "C", "D"]
freq = {x: 0 for x in L}

def reorder():
    global L
    L = sorted(L, key=lambda x: (-freq[x], L.index(x)))

# Access A ten times
for _ in range(10):
    freq["A"] += 1
    reorder()

print("After 10 accesses to A:")
print("freq =", freq)
print("list =", L)
print()

# Now access D repeatedly
for i in range(1, 6):
    freq["D"] += 1
    reorder()

    print(f"After {i} accesses to D:")
    print("freq =", freq)
    print("list =", L)
    print()
```

```python
"""
Even after several accesses to D, the old frequency of A keeps A near the front.
This "long memory" is exactly why frequency count can perform much worse than Move-to-Front when access patterns change.
"""
```

```python
"""
Marked pages are protected.
A page becomes marked when it is requested.
Once every cached page has been requested during the current phase, the phase ends and all marks are cleared.
Evictions are chosen uniformly from the unmarked pages only.

x* = page x is marked
x  = page x is unmarked

Only unmarked pages may be evicted.
When all pages become marked, all marks are cleared and a new phase begins.
"""
import random

class Page:
    def __init__(self, value):
        self.value = value
        self.mark = False

    def __repr__(self):
        return f"{self.value}{'*' if self.mark else ''}"


def randomized_marking(requests, cache_size):
    cache = []
    faults = 0

    for req in requests:
        print(f"\nRequest: {req}")
        hit = False
        for page in cache:
            if page.value == req:
                page.mark = True
                hit = True
                print("HIT")
                break

        if not hit:
            faults += 1
            print("MISS")

            # Cache not full yet
            if len(cache) < cache_size:
                cache.append(Page(req))
                cache[-1].mark = True

            else:
                # If all pages marked, unmark them
                if all(page.mark for page in cache):
                    print("All pages marked -> unmarking")
                    for page in cache:
                        page.mark = False

                # Choose random unmarked page
                candidates = [p for p in cache if not p.mark]
                victim = random.choice(candidates)

                print(f"Evicting {victim.value}")

                cache.remove(victim)

                new_page = Page(req)
                new_page.mark = True
                cache.append(new_page)

        print("Cache:", cache)

    print("\nTotal faults:", faults)
```

```python
"""
(Least Frequency used) OPT (optimal offline)
LFU remembers all past history. If page frequencies were high in the distant past,
LFU can become "stuck" keeping obsolete pages forever. An adversary can exploit this by:
Making some pages extremely frequent early,
Then switching permanently to a completely different working set.

OPT adapts immediately, while LFU keeps making misses, yielding a ratio proportional to n/k.
OPT sees the future.

After the first cycle through
k+1,…,2k

it keeps exactly these pages in cache.
The first k requests miss, but afterward all requests hit.




FIFO can lose at most k times per phase.
OPT must pay at least once every two phases.
Therefore FIFO's cost is at most a factor of 2k larger.

Hence the competitive ratio of FIFO is
O(k)
"""
```

```python
"""
(r, p) release time, processing time

Compute SRPT.
Sort jobs by SRPT completion time.
Run them nonpreemptively in that order.
"""
import heapq


def srpt_completion_times(tasks):
    n = len(tasks)

    arrivals = sorted(
        [(r, p, i) for i, (r, p) in enumerate(tasks)]
    )

    completion = [0] * n
    heap = []
    t = 0
    k = 0

    while k < n or heap:
        if not heap:
            t = arrivals[k][0]

        while k < n and arrivals[k][0] <= t:
            r, p, idx = arrivals[k]
            heapq.heappush(heap, (p, idx))
            k += 1

        remaining, idx = heapq.heappop(heap)

        next_arrival = (
            arrivals[k][0] if k < n else float("inf")
        )

        if t + remaining <= next_arrival:
            t += remaining
            completion[idx] = t
        else:
            work = next_arrival - t
            remaining -= work
            t = next_arrival
            heapq.heappush(heap, (remaining, idx))

    return completion
```

```python
def completion_time_schedule(tasks):
    cp = srpt_completion_times(tasks)

    # renumber tasks by SRPT completion order
    order = sorted(range(len(tasks)), key=lambda i: cp[i])

    t = 0
    completion = {}

    for i in order:
        r, p = tasks[i]

        t = max(t, r)
        t += p

        completion[i] = t

    return completion
```
