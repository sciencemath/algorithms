```python
"""
A(x) = a_0 + a_1x + a_2x^2 + ... + a_nx^n.

Suppose we divide A(x) by (x - x_0), obtaining
A(x) = q(x)(x - x_0) + r,

where
q(x) = b_0 + b_1x + b_2x^2 + ... + b_(n-1)x^(n-1)

and r is the remainder. Since A(x_0) = r, we need to compute both r and the coefficients b_i efficiently.

Expanding q(x)(x - x_0) gives

A(x) = (-x_0b_0)
     + (b_0 - x_0b_1)x
     + (b_1 - x_0b_2)x^2
     + ...
     + (b_(n-2) - x_0b_(n-1))x^(n-1)
     + b_(n-1)x^n
     + r.

Matching coefficients with A(x) yields

a_n = b_(n-1),

a_(n-1) = b_(n-2) - x_0b_(n-1),

a_(n-2) = b_(n-3) - x_0b_(n-2),

...

a_1 = b_0 - x_0b_1,

a_0 = r - x_0b_0.

These equations allow us to compute the quotient coefficients from highest degree to lowest:

b_(n-1) = a_n,
b_(n-2) = a_(n-1) + x_0b_(n-1),
b_(n-3) = a_(n-2) + x_0b_(n-2),

...

b_i = a_(i+1) + x_0b_(i+1).

Finally, once b_0 has been computed,
r = a_0 + x_0b_0.

Algorithm:
1. Set b_(n-1) = a_n.
2. For i = n-2, n-3, ..., 0:
      b_i = a_(i+1) + x_0b_(i+1).
3. Compute
      r = a_0 + x_0b_0.
4. Return q(x) and r.

The algorithm performs one pass through the coefficients of A(x). Each iteration requires only one multiplication and one addition, so the total running time is Θ(n). This procedure is known as synthetic division and produces both the quotient polynomial q(x) and the remainder r = A(x_0) in linear time.
"""
```

```python
"""
To evaluate a polynomial A(x) at a point x0 in Θ(n) time, we divide A(x) by (x - x0):

A(x) = q(x)(x - x0) + r.

By the Remainder Theorem, substituting x = x0 gives

A(x0) = q(x0)(x0 - x0) + r = r,

so the remainder r is the value of the polynomial at x0.

Consider the example

A(x) = x^2 + 3x + 2

and let

x0 = 1.

The coefficients are

a2 = 1, a1 = 3, a0 = 2.

Since A(x) has degree 2, the quotient q(x) has degree 1:

q(x) = b1x + b0.

Using synthetic division, begin with the highest-degree coefficient:

b1 = a2 = 1.

Next compute

b0 = a1 + x0*b1
   = 3 + (1)(1)
   = 4.

Finally compute the remainder:

r = a0 + x0*b0
  = 2 + (1)(4)
  = 6.

Therefore,

q(x) = x + 4

and

r = 6.

Checking:

(x + 4)(x - 1) + 6
= x^2 + 3x - 4 + 6
= x^2 + 3x + 2,

which matches the original polynomial. Also,

A(1) = 1^2 + 3(1) + 2 = 6,

which equals the remainder.

In general, for

A(x) = a0 + a1x + ... + anx^n,

the algorithm is:

b[n-1] = a[n]

for i = n-2 down to 0:
    b[i] = a[i+1] + x0*b[i+1]

r = a[0] + x0*b[0]

This algorithm performs one multiplication and one addition for each coefficient, requiring only a single pass through the coefficients.
Therefore, the running time is Θ(n).
"""
```

```python
import cmath

def fft(A):
    n = len(A)

    if n == 1:
        return A

    A_even = A[0::2]
    A_odd  = A[1::2]

    y_even = fft(A_even)
    y_odd  = fft(A_odd)

    # Primitive nth root of unity
    omega_n = cmath.exp(2j * cmath.pi / n)

    y = [0] * n
    omega = 1

    for k in range(n // 2):
        t = omega * y_odd[k]

        y[k] = y_even[k] + t
        y[k + n // 2] = y_even[k] - t

        omega *= omega_n

    return y
```

```python
"""
Example: FFT on A = [1, 2, 3, 4], representing A(x) = 1 + 2x + 3x^2 + 4x^3.

Step 1: Split into even and odd coefficients
A_even = [1, 3]
A_odd  = [2, 4]

So:
A(x) = A_even(x^2) + x * A_odd(x^2)

Step 2: Compute FFT(A_even)
A_even = [1, 3]

Split again:
[1] and [3]

Base cases:
FFT([1]) = [1]
FFT([3]) = [3]

Combine (n = 2, ω₂ = -1):
y_even[0] = 1 + 3 = 4
y_even[1] = 1 - 3 = -2

So:
FFT([1, 3]) = [4, -2]

Step 3: Compute FFT(A_odd)
A_odd = [2, 4]

Split again:
[2] and [4]

Base cases:
FFT([2]) = [2]
FFT([4]) = [4]

Combine (ω₂ = -1):
y_odd[0] = 2 + 4 = 6
y_odd[1] = 2 - 4 = -2

So:
FFT([2, 4]) = [6, -2]

Step 4: Combine results (n = 4, ω₄ = i)
y_even = [4, -2]
y_odd  = [6, -2]

k = 0:
t = 1 * 6 = 6
y[0] = 4 + 6 = 10
y[2] = 4 - 6 = -2

k = 1:
t = i * (-2) = -2i
y[1] = -2 - 2i
y[3] = -2 + 2i

Final result:
FFT([1,2,3,4]) = [10, -2-2i, -2, -2+2i]

This corresponds to evaluating A(x) at:
x = 1, i, -1, -i
"""
```

```python
"""
Inverse FFT (Θ(n log n))

FFT = rotate forward around unit circle
IFFT = rotate backward (complex conjugate)
scaling by 1/n fixes magnitude
"""
def ifft(a):
    n = len(a)

    # Step 1: conjugate input
    a_conj = [x.conjugate() for x in a]

    # Step 2: apply FFT
    y = fft(a_conj)

    # Step 3: conjugate again and scale
    return [x.conjugate() / n for x in y]
```

```python
"""
Think of the FFT as answering the question:
"How much of each frequency exists in the input signal?"

The DFT directly computes every frequency by summing all n inputs for every output frequency, requiring n^2 operations.

The FFT notices that:
ω^(k(j+n/2))_n = −ω^(kj)_n
which means half of the DFT terms are closely related to the other half. Instead of computing everything from scratch, the FFT:

Splits the input into smaller pieces.
Solves the smaller DFTs.
Recombines them using butterflies.

Stage 0:  8 individual numbers
Stage 1:  4 groups of 2
Stage 2:  2 groups of 4
Stage 3:  1 group of 8

For n=8:

Stage 1 combines pairs (size 2 groups).
Stage 2 combines those results into size 4 groups.
Stage 3 combines those results into one size 8 group.

The FFT repeatedly performs these add/subtract operations because addition and subtraction are much cheaper than
recomputing the large sums appearing in the DFT.

      u
     / \
    /   \
   +     -
    \   /
     \ /
      v

TOP: u + v
BOTTOM: u - v
The top output captures the part where the two inputs reinforce each other (even-frequency contribution),
while the bottom output captures the part where they cancel each other (odd-frequency contribution).

The Fast Fourier Transform (FFT) computes the same result as the Discrete Fourier Transform (DFT) but much more efficiently.
Instead of directly evaluating every DFT sum, which requires O(n²) operations, the FFT uses a divide-and-conquer strategy
that reduces the complexity to O(n log n). For an 8-point FFT, there are log₂(8)=3 stages of butterflies.
A butterfly takes two inputs u and v and produces the outputs u+v and u−v.
The intuition is that the DFT contains many repeated calculations, and the FFT exploits symmetries in the roots
of unity to reuse intermediate results. At each stage, progressively larger groups of data are combined.
Stage 1 combines groups of size 2, Stage 2 combines groups of size 4, and Stage 3 combines groups of size 8.
The top butterfly output (u+v) captures the component where the inputs reinforce each other,
while the bottom output (u−v) captures the component where they cancel each other. Twiddle factors ω₈ᵏ are complex roots
of unity that introduce the phase shifts required by the DFT formula when merging the smaller transforms back together.

For the input vector [0, 2, 3, -1, 4, 5, 7, 9], Stage 0 consists of the original inputs:

[0, 2, 3, -1, 4, 5, 7, 9]

Stage 1 performs butterflies on adjacent pairs:

(0, 2)  -> (0+2, 0-2)   = (2, -2)
(3, -1) -> (3+(-1), 3-(-1)) = (2, 4)
(4, 5)  -> (4+5, 4-5)   = (9, -1)
(7, 9)  -> (7+9, 7-9)   = (16, -2)

After Stage 1 the wire values are:

[2, -2, 2, 4, 9, -1, 16, -2]

Stage 2 combines values separated by two positions. The twiddle factors are ω₈⁰=1 and ω₈²=i.

Butterfly (0,2):
Inputs: (2,2)
Output: (2+2, 2-2) = (4,0)

Butterfly (1,3):
Inputs: (-2,4)
Twiddle multiplication: 4·ω₈² = 4i
Output:
(-2+4i, -2-4i)

Butterfly (4,6):
Inputs: (9,16)
Output:
(25,-7)

Butterfly (5,7):
Inputs: (-1,-2)
Twiddle multiplication:
(-2)·ω₈² = -2i
Output:
(-1-2i, -1+2i)

After Stage 2 the wire values are:

[4, -2+4i, 0, -2-4i, 25, -1-2i, -7, -1+2i]

Stage 3 combines values separated by four positions. The twiddle factors are:

ω₈⁰ = 1
ω₈¹ = (√2/2)(1+i)
ω₈² = i
ω₈³ = (-1+i)/√2

Butterfly (0,4):
Inputs: (4,25)
Output:
(29,-21)

Butterfly (1,5):
Inputs: (-2+4i, -1-2i)

Twiddled lower input:
(-1-2i)ω₈ = (1-3i)/√2

Output:
(-2+4i + (1-3i)/√2,
 -2+4i - (1-3i)/√2)

≈ (-1.2929+1.8787i,
   -2.7071+6.1213i)

Butterfly (2,6):
Inputs: (0,-7)

Twiddled lower input:
(-7)i = -7i

Output:
(-7i, 7i)

Butterfly (3,7):
Inputs: (-2-4i, -1+2i)

Twiddled lower input:
(-1+2i)ω₈³ = (-1-3i)/√2

Output:
(-2-4i + (-1-3i)/√2,
 -2-4i - (-1-3i)/√2)

≈ (-2.7071-6.1213i,
   -1.2929-1.8787i)

The final FFT output is therefore:

X₀ = 29
X₁ = -1.2929 + 1.8787i
X₂ = -7i
X₃ = -2.7071 - 6.1213i
X₄ = -21
X₅ = -2.7071 + 6.1213i
X₆ = 7i
X₇ = -1.2929 - 1.8787i

or in vector form:

[29,
 -1.2929+1.8787i,
 -7i,
 -2.7071-6.1213i,
 -21,
 -2.7071+6.1213i,
 7i,
 -1.2929-1.8787i]

The key insight is that instead of computing every DFT coefficient directly from all eight inputs,
the FFT repeatedly combines smaller partial transforms using simple additions, subtractions,
and a small number of multiplications by roots of unity (twiddle factors).
This reuse of intermediate computations is what reduces the running time from O(n²) to O(n log n).
"""
```

```python
def bit_reverse_permutation(a, n):
    """
    In-place bit-reversal permutation of array a of length n = 2^b.
    Uses bit_reverse_of(k, b) which returns the b-bit reversal of k.
    """

    b = (n - 1).bit_length()  # since n = 2^b

    for k in range(n):
        rk = bit_reverse_of(k, b)

        # swap only if needed (avoid double swapping)
        if k < rk:
            a[k], a[rk] = a[rk], a[k]

def bit_reverse_of(k, b):
    r = 0
    for _ in range(b):
        r = (r << 1) | (k & 1)
        k >>= 1
    return r
```
