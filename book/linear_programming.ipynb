```python
"""
The Fundamental Theorem of Linear Programming usually refers to:
If a linear program has an optimal solution, then it has an optimal solution at an extreme point (vertex) of the feasible region.

In other words, when optimizing a linear objective over a polyhedron, you only need to look at the vertices.
"""
```

```python
"""
The feasible region of a linear program is a convex polyhedron.
If an optimal solution is not already at a vertex, then it lies on a line segment
connecting two other feasible points. Because the objective function is linear,
it must take the same optimal value at both endpoints of that segment.
Repeating this argument eventually leads to an optimal vertex.
"""
```

```python
"""
To formulate the dual, first observe that the primal linear program is in the standard minimization form:

minimize cᵀx
subject to Ax ≥ b,
x ≥ 0.

The corresponding dual of a minimization problem with "≥" constraints and nonnegative variables is:

maximize bᵀy
subject to Aᵀy ≤ c,
y ≥ 0.

For the given problem,

A = [[-2, 8, 0, 10],
     [ 5, 2, 0,  0],
     [ 3,-5,10, -2]]

b = [50, 100, 25]ᵀ

c = [1, 1, 1, 1]ᵀ.

Introducing dual variables y₁, y₂, and y₃ corresponding to the three primal constraints, the dual objective function is obtained from bᵀy:

maximize 50y₁ + 100y₂ + 25y₃.

The dual constraints come from Aᵀy ≤ c. Using the columns of A, we obtain:

-2y₁ + 5y₂ + 3y₃ ≤ 1,
8y₁ + 2y₂ - 5y₃ ≤ 1,
10y₃ ≤ 1,
10y₁ - 2y₃ ≤ 1.

Since the primal is a minimization problem with constraints of the form Ax ≥ b and x ≥ 0, the dual variables must satisfy y₁, y₂, y₃ ≥ 0.

Therefore, the dual linear program is:

maximize 50y₁ + 100y₂ + 25y₃

subject to

-2y₁ + 5y₂ + 3y₃ ≤ 1,
8y₁ + 2y₂ - 5y₃ ≤ 1,
10y₃ ≤ 1,
10y₁ - 2y₃ ≤ 1,

y₁, y₂, y₃ ≥ 0.

A quick way to remember this conversion is that a primal minimization problem in the form Ax ≥ b with x ≥ 0
becomes a dual maximization problem with constraints A^Ty ≤ c and y ≥ 0. The right-hand side vector b becomes
the coefficients of the dual objective function, and the primal objective coefficients c become the right-hand
sides of the dual constraints.
"""
```

```python
"""
To directly construct the dual of an arbitrary linear program, it is not necessary to first convert the primal
into standard form. Instead, we can apply a set of correspondence rules between
primal and dual variables, constraints, and objective functions.

Suppose the primal is either a maximization or minimization problem with a mixture of ≤, ≥, and = constraints,
and variables that may be nonnegative, nonpositive, or unrestricted in sign.
To form the dual, first create one dual variable for each primal constraint.
The coefficient matrix of the dual is the transpose of the primal coefficient matrix,
so rows of the primal become columns of the dual and vice versa.
The right-hand side constants of the primal constraints become the coefficients of the dual objective function,
while the coefficients of the primal objective function become the right-hand side constants of the dual constraints.

The signs of the dual variables are determined by the types of primal constraints.
If the primal is a maximization problem, a primal constraint of the form ≤ produces a nonnegative dual variable,
a constraint of the form ≥ produces a nonpositive dual variable, and an equality constraint produces a dual variable
that is unrestricted in sign. For a minimization problem, these sign rules are reversed.

Similarly, the form of each dual constraint is determined by the sign restriction on the corresponding primal variable.
A primal variable constrained to be nonnegative produces a dual constraint of the appropriate inequality direction
(≥ for a dual of a maximization problem, ≤ for a dual of a minimization problem).
A primal variable constrained to be nonpositive reverses that inequality direction.
A primal variable that is unrestricted in sign produces an equality constraint in the dual.

Thus, the general procedure for directly forming the dual is:

1. Create one dual variable for each primal constraint.
2. Transpose the coefficient matrix A to obtain Aᵀ.
3. Use the primal right-hand side vector b as the coefficients of the dual objective function.
4. Use the primal objective coefficient vector c as the right-hand sides of the dual constraints.
5. Determine the sign restrictions of dual variables from the primal constraint types.
6. Determine the dual constraint types from the primal variable sign restrictions.

These rules allow the dual of any linear program to be written directly, without first converting the primal into standard form.

| Primal Constraint | Dual Variable      |
| ----------------- | ------------------ |
| <=                | (y_i >= 0)         |
| >=                | (y_i <= 0)         |
| =                 | (y_i) unrestricted |


| Primal Variable    | Dual Constraint |
| ------------------ | --------------- |
| (x_j >= 0)         | >=              |
| (x_j <= 0)         | <=              |
| (x_j) unrestricted | =               |

"""
```

```python
"""
The dual of the maximum-flow linear program can be interpreted as a minimum-cut problem.
The dual variables p_u correspond to vertices and indicate which side of a cut each vertex belongs to.
We may think of p_u = 1 for vertices in the source side S and p_u = 0 for vertices in the sink side T, with s ∈ S and t ∈ T.
The dual variables y_uv correspond to edges and indicate whether an edge crosses the cut.
The dual constraints force y_uv to be positive whenever an edge crosses from one side of the cut to the other.
Since the dual objective is minimize Σ_(u,v) c(u,v)y_uv, only crossing edges contribute to the objective value.

Consequently, the objective equals the total capacity of the cut.
Therefore the dual LP searches for a partition of the vertices that separates the source and sink while minimizing
the total capacity of crossing edges. This is exactly the minimum-cut problem. By strong duality,
the optimal value of the maximum-flow LP equals the optimal value of the minimum-cut LP, which is the Max-Flow Min-Cut Theorem.
"""
```

```python
"""
To form the dual of the minimum-cost-flow problem, introduce a nonnegative dual variable y_uv for each capacity
constraint f_uv ≤ c(u,v) and an unrestricted dual variable p_u for each flow-conservation equation.
The resulting dual is
maximize d(p_s − p_t) − Σ_(u,v) c(u,v)y_uv

subject to
p_u − p_v − y_uv ≤ a(u,v) for every edge (u,v),
y_uv ≥ 0 for every edge (u,v).

The dual has a natural graph interpretation. The variables p_u represent potentials (or prices) assigned to vertices.
The constraint p_u − p_v − y_uv ≤ a(u,v) says that the difference in potential between adjacent vertices cannot exceed
the cost of the edge unless we pay for a violation through y_uv. Each unit of violation on edge (u,v) incurs a penalty
proportional to its capacity c(u,v). The objective attempts to maximize the potential difference between the
source and sink while minimizing these penalties. If capacities are infinite, then y_uv must be zero,
and the constraints reduce to p_u − p_v ≤ a(u,v), which are exactly the shortest-path inequalities.
Thus the dual can be viewed as a shortest-path-type problem involving vertex potentials and edge-capacity penalties.
"""
```

```python
"""
The result from Chapter 24 that can be interpreted as weak duality is the theorem stating that the value of any
feasible flow is at most the capacity of any s-t cut.

In notation:
|f| ≤ c(S,T)

for every feasible flow f and every cut (S,T). This is exactly the weak duality property of linear programming.
The maximum-flow problem is the primal LP and the minimum-cut problem is its dual LP.
Weak duality states that the objective value of any feasible primal solution cannot exceed the objective value
of any feasible dual solution. Translating this statement to flows and cuts gives "flow value ≤ cut capacity."
The Max-Flow Min-Cut Theorem, which states that the maximum flow value equals the minimum cut capacity,
can then be viewed as the corresponding strong duality theorem.
"""
```

```python
"""
Let P = {x : Ax ≤ b} be the feasible region and suppose x* is an optimal solution to maximize c^T x over P.
If x* is already an extreme point, then the theorem is proved. Otherwise, x* is not an extreme point,
so there exist distinct feasible points u and v and a number 0 < λ < 1 such that x* = λu + (1 − λ)v.
Since the objective function is linear,

c^T x* = c^T(λu + (1 − λ)v)
       = λ(c^T u) + (1 − λ)(c^T v).

Because x* is optimal, c^T x* equals the optimal value z*. Since u and v are feasible, c^T u ≤ z* and c^T v ≤ z*.
The weighted average λ(c^T u) + (1 − λ)(c^T v) can equal z* only if c^T u = z* and c^T v = z*.
Therefore u and v are also optimal solutions. If either u or v is an extreme point, we are done.
Otherwise, repeat the argument. Continuing in this way eventually yields an optimal extreme point of the feasible region.
Hence every linear program with an optimal solution has an optimal solution at a vertex of the feasible region.
"""
```
