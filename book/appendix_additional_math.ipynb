# Summation Proof Formula Reference

## 1. Big-O Summation Property

Goal:

$$
\sum_{k=1}^{n} O(f_k(i))
=
O\left(\sum_{k=1}^{n} f_k(i)\right)
$$

Definition of Big-O:

$$
g_k(i) \leq c_k f_k(i)
$$

Choose the largest constant:

$$
c = \max(c_1,c_2,\dots,c_n)
$$

Then:

$$
g_k(i) \leq c f_k(i)
$$

Sum both sides:

$$
\sum_{k=1}^{n} g_k(i)
\leq
\sum_{k=1}^{n} c f_k(i)
$$

Use linearity of summation:

$$
\sum_{k=1}^{n} c f_k(i)
=
c\sum_{k=1}^{n} f_k(i)
$$

Therefore:

$$
\sum_{k=1}^{n} g_k(i)
\leq
c\sum_{k=1}^{n} f_k(i)
$$

So:

$$
\boxed{
\sum_{k=1}^{n} O(f_k(i))
=
O\left(\sum_{k=1}^{n} f_k(i)\right)
}
$$

# Generalized DeMorgan's Laws

For a finite collection of sets:

$$
A_1,A_2,\dots,A_n
$$


## First DeMorgan Law

$$
\boxed{
\overline{\bigcup_{i=1}^{n}A_i}
=
\bigcap_{i=1}^{n}\overline{A_i}
}
$$


Proof:

Take an arbitrary element x:

$$
x\in\overline{\bigcup_{i=1}^{n}A_i}
$$

means:

$$
x\notin\bigcup_{i=1}^{n}A_i
$$

By the definition of union:

$$
x\notin A_i
$$

for every i.

Therefore:

$$
x\in\overline{A_i}
$$

for every i.

Hence:

$$
x\in\bigcap_{i=1}^{n}\overline{A_i}
$$

Therefore:

$$
\boxed{
\overline{\bigcup_{i=1}^{n}A_i}
=
\bigcap_{i=1}^{n}\overline{A_i}
}
$$



## Second DeMorgan Law

$$
\boxed{
\overline{\bigcap_{i=1}^{n}A_i}
=
\bigcup_{i=1}^{n}\overline{A_i}
}
$$


Proof:

Take an arbitrary element x:

$$
x\in\overline{\bigcap_{i=1}^{n}A_i}
$$

means:

$$
x\notin\bigcap_{i=1}^{n}A_i
$$

Therefore x is not in at least one of the sets:

$$
x\notin A_i
$$

for some i.

Thus:

$$
x\in\overline{A_i}
$$

for some i.

Therefore:

$$
x\in\bigcup_{i=1}^{n}\overline{A_i}
$$

Hence:

$$
\boxed{
\overline{\bigcap_{i=1}^{n}A_i}
=
\bigcup_{i=1}^{n}\overline{A_i}
}
$$

# Set Theory Proof Notes

## Inclusion-Exclusion Principle

For n sets:

$$
|A_1\cup A_2\cup\cdots\cup A_n|
$$

is:

$$
\sum_i |A_i|
-\sum_{i<j}|A_i\cap A_j|
+\sum_{i<j<k}|A_i\cap A_j\cap A_k|
-\cdots
+(-1)^{n-1}|A_1\cap\cdots\cap A_n|
$$


The reason for alternating signs is that adding individual sets counts overlaps multiple times. Pair intersections are subtracted to remove double counting, triple intersections are added back because they were removed too many times, and the pattern continues.

# Functions and Cardinality


## (a) Injective Functions

If

$$
f:A\rightarrow B
$$

is injective, then every element of A maps to a different element of B.

Different inputs produce different outputs:

$$
a_1\neq a_2
\Longrightarrow
f(a_1)\neq f(a_2)
$$

Since no element of B can be reused,

$$
|A|
\le
|B|
$$

Therefore:

$$
\boxed{|A|\le|B|}
$$



## Surjective Functions

If

$$
f:A\rightarrow B
$$

is surjective, then every element of B has at least one preimage.

Definition:

$$
\forall b\in B,\;
\exists a\in A
\text{ such that }
f(a)=b
$$

Therefore A must contain at least as many elements as B.

Hence:

$$
\boxed{|A|\ge|B|}
$$



---

# (b) Function f(x)=x+1

On

$$
\mathbb N
$$

Injective:

$$
x_1+1=x_2+1
\Rightarrow
x_1=x_2
$$

Not surjective because:

$$
x+1=1
$$

has no solution in

$$
\mathbb N
$$

Therefore:

$$
\boxed{\text{Not bijective on }\mathbb N}
$$



On

$$
\mathbb Z
$$

For every integer y,

$$
x=y-1
$$

exists.

Therefore:

Injective + Surjective

$$
\boxed{\text{Bijective on }\mathbb Z}
$$

# (c) Inverse Relation

Definition:

$$
R^{-1}
=
\{
(b,a)
:
(a,b)\in R
\}
$$

Every ordered pair is reversed.

If R is a bijective function, this agrees with the ordinary inverse function.

# (d) Bijection Between Z and Z×Z

Map integers to naturals:

$$
g(z)=
\begin{cases}
2z,&z\ge0\\
-2z-1,&z<0
\end{cases}
$$

Use Cantor pairing:

$$
\pi(a,b)
=
\frac{(a+b)(a+b+1)}2+b
$$

Combine the two mappings to obtain a bijection:

$$
\boxed{
\mathbb Z
\leftrightarrow
\mathbb Z\times\mathbb Z
}
$$

This works because both sets are countably infinite.

````python
"""
Concrete Example
Suppose

(2,−1)

Convert:
2 → 4
-1 → 1

Pair:
(4,1)

Cantor pairing gives one natural number.
Reverse the process to recover
(2,-1)

Nothing is lost.
"""
````

# Graph Theory Proofs

## (a) Handshaking Lemma

Given an undirected graph:

$$
G=(V,E)
$$

Every edge contributes exactly 1 to the degree of each endpoint.

Thus each edge contributes:

$$
2
$$

to the total degree.

Therefore:

$$
\boxed{
\sum_{v\in V}\deg(v)=2|E|
}
$$

Since

$$
2|E|
$$

is even, the sum of all vertex degrees is always even.

---

## (b) Every Path Contains a Simple Path

Suppose a path contains a repeated vertex.

Example:

$$
A\rightarrow B\rightarrow C\rightarrow B\rightarrow D
$$

The repeated section:

$$
B\rightarrow C\rightarrow B
$$

forms a cycle.

Removing this cycle leaves:

$$
A\rightarrow B\rightarrow D
$$

Repeating this process eventually produces a simple path.

Similarly, every cycle contains a simple cycle by removing repeated vertices until only one repetition (start/end) remains.

---

## (c) Connected Graph

A tree with

$$
|V|
$$

vertices always has

$$
|V|-1
$$

edges.

Every connected graph contains a spanning tree.

Adding edges only increases the number of edges.

Therefore:

$$
\boxed{
|E|\ge |V|-1
}
$$

---

## (d) Reachability

### Undirected Graph

Reachability satisfies:

Reflexive:

$$
v
$$

can always reach itself.

Symmetric:

If

$$
u
$$

can reach

$$
v
$$

then

$$
v
$$

can reach

$$
u
$$

because every edge is bidirectional.

Transitive:

If

$$
u
$$

can reach

$$
v
$$

and

$$
v
$$

can reach

$$
w
$$

then

$$
u
$$

can reach

$$
w
$$

Therefore reachability is an equivalence relation.

For directed graphs, symmetry may fail.

Example:

$$
A\rightarrow B
$$

A reaches B, but B does not necessarily reach A.

Thus reachability is not generally an equivalence relation.

---

## (e) Hypergraph Representation

Let the left side of a bipartite graph represent vertices:

$$
V
$$

Let the right side represent hyperedges:

$$
E
$$

Draw an edge whenever a vertex belongs to a hyperedge.

If

$$
e=\{v_1,v_2,v_3\}
$$

then connect:

$$
v_1-e
$$

$$
v_2-e
$$

$$
v_3-e
$$

Every edge connects one vertex node to one hyperedge node, making the graph bipartite.

| Problem                | Main idea                                     | Easy way to remember                               |
| ---------------------- | --------------------------------------------- | -------------------------------------------------- |
| Handshaking Lemma      | Every edge has two endpoints                  | Every handshake is counted twice                   |
| Simple path            | Remove loops                                  | Erase repeated vertices                            |
| Connected graph        | Minimum connected graph is a tree             | Trees have (V-1) edges                             |
| Reachability           | Undirected paths work both ways               | Directed paths do not                              |
| Hypergraph → Bipartite | Vertices on one side, hyperedges on the other | Connect a vertex to the hyperedges that contain it |

# Graph Coloring Proofs

## (a) Every Tree is 2-Colorable

Choose any vertex as the root.

Assign colors by depth:

- Even depth → Color 1
- Odd depth → Color 2

Every edge connects vertices whose depths differ by exactly one.

Therefore adjacent vertices always receive different colors.

Hence

$$
\boxed{\text{Every tree is 2-colorable}}
$$

---

## (b) Equivalent Statements

The following are equivalent:

1.

$$
G
$$

is bipartite.

2.

$$
G
$$

is 2-colorable.

3.

$$
G
$$

contains no odd cycles.

Proof:

- Bipartite ⇒ 2-colorable by coloring the two partitions.
- 2-colorable ⇒ bipartite by letting each color class be a partition.
- 2-colorable ⇒ no odd cycles because an odd cycle forces two adjacent vertices to receive the same color.
- No odd cycles ⇒ 2-colorable by performing BFS and coloring vertices according to even or odd distance from the starting vertex.

Therefore

$$
\boxed{
(1)\iff(2)\iff(3)
}
$$

---

## (c) Coloring with d+1 Colors

Let

$$
d=\max\deg(v)
$$

Color vertices one at a time.

When coloring any vertex,

at most

$$
d
$$

neighbors already have colors.

Among

$$
d+1
$$

colors,

at least one color is unused.

Therefore

$$
\boxed{
G
\text{ can always be colored with }
d+1
\text{ colors}
}
$$

---

## (d) Sparse Graph Coloring

Suppose

$$
|E|=O(|V|)
$$

The average degree is

$$
\frac{2|E|}{|V|}
=
O(1)
$$

Vertices with degree greater than

$$
\sqrt{|V|}
$$

are few because each contributes many edges.

There are only

$$
O(\sqrt{|V|})
$$

such vertices.

The remaining vertices have degree at most

$$
\sqrt{|V|}
$$

Applying part (c),

they require at most

$$
\sqrt{|V|}+1
$$

colors.

Therefore

$$
\boxed{
G
\text{ can be colored with }
O(\sqrt{|V|})
\text{ colors}
}
$$

| Bipartite ⇔ 2-colorable | Two partitions are exactly two colors.                                                      |

# Friendship Graph Theorems

## (a) Equal Degrees

**Theorem**

Every simple undirected graph with

$$
n\ge2
$$

vertices contains two vertices of the same degree.

Proof:

Every degree is in

$$
\{0,1,\ldots,n-1\}.
$$

A graph cannot simultaneously contain a vertex of degree

$$
0
$$

and a vertex of degree

$$
n-1,
$$

since a vertex adjacent to every other vertex prevents any isolated vertex.

Thus there are at most

$$
n-1
$$

possible degree values for

$$
n
$$

vertices.

By the Pigeonhole Principle,

$$
\boxed{\text{two vertices have the same degree}.}
$$

---

## (b) Ramsey's Theorem

**Theorem**

Every graph on six vertices contains either

- a triangle, or
- an independent set of three vertices.

Proof:

Choose any vertex.

Among its five neighbors,

at least three are adjacent or at least three are nonadjacent.

If two of the adjacent vertices are connected,

they form a triangle.

Otherwise the three vertices form an independent set.

Thus

$$
\boxed{R(3,3)=6.}
$$

---

## (c) Maximum Cut

**Theorem**

Every finite graph has a partition

$$
V=A\cup B
$$

such that every vertex has at least half of its neighbors in the opposite part.

Proof:

Choose a partition maximizing the number of crossing edges.

If some vertex has more neighbors on its own side than across,

moving it increases the number of crossing edges,

contradicting maximality.

Therefore every vertex has at least half of its neighbors across the partition.

---

## (d) Dirac's Theorem

**Theorem**

If every vertex satisfies

$$
\deg(v)\ge\frac{|V|}{2},
$$

then the graph contains a Hamiltonian cycle.

Traversing this Hamiltonian cycle provides a circular seating arrangement in which every person sits between two friends.

Therefore

$$
\boxed{\text{the seating arrangement always exists}.}
$$

# Tree Bisection Proofs

## (a) One-Edge Separator

**Theorem**

Every binary tree with

$$
n
$$

vertices contains an edge whose removal partitions the tree into two connected components

$$
A,B
$$

such that

$$
|A|
\le
\frac{3n}{4},
\qquad
|B|
\le
\frac{3n}{4}.
$$

Proof:

Starting at the root, repeatedly move to a child whose subtree contains more than

$$
\frac{3n}{4}
$$

vertices.

Subtree sizes strictly decrease.

Eventually we stop at a subtree

$$
T
$$

whose size satisfies

$$
\frac n4
\le
|T|
\le
\frac{3n}{4}.
$$

Removing the edge above

$$
T
$$

creates two connected components of sizes

$$
|T|
$$

and

$$
n-|T|,
$$

both at most

$$
\frac{3n}{4}.
$$

---

## (b) Optimality

Construct a binary tree where one child subtree contains

$$
\frac{3n}{4}
$$

vertices.

Every single edge either isolates a very small subtree or leaves the large subtree intact.

Thus every partition has one side of size at least

$$
\frac{3n}{4},
$$

showing the constant cannot be improved.

---

## (c) Exact Bisection

Repeatedly apply the separator theorem to the largest remaining component.

Each cut reduces the largest component by a factor of at least

$$
\frac34.
$$

After

$$
k
$$

cuts,

the largest remaining component has size

$$
\left(\frac34\right)^k n.
$$

Choose

$$
k
=
\left\lceil
\log_{4/3}n
\right\rceil
=
O(\log n).
$$

The remaining components have constant size.

Combine them to obtain two vertex sets of sizes

$$
\left\lfloor\frac n2\right\rfloor
$$

and

$$
\left\lceil\frac n2\right\rceil.
$$

Therefore only

$$
O(\log n)
$$

edges are required.
