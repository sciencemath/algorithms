````python
"""
When the input contains many repeated points, choosing the initial centers uniformly at random may select multiple centers
that are identical or very close together. This produces poor initial clusters because several centers represent the same region of the data.

A better initialization procedure is to choose the first center uniformly at random from all points.
Then, for each remaining point, compute its distance to the nearest center that has already been chosen.
Select the next center with probability proportional to the square of this distance.
Consequently, points that are far away from all existing centers are much more likely to be selected,
while duplicate or nearby points have a very small probability of being chosen. Repeat this process until k centers have been selected.

This initialization spreads the centers throughout the data, maximizes the number of distant centers chosen,
and typically leads to better clustering results and faster convergence of Lloyd's algorithm.
"""
````

````python
"""
Show how to find an optimal k-clustering in polynomial time when there is just one attribute (d = 1):
(points are on a line vs a plane)

The reason this works is that in one dimension every optimal cluster is a contiguous block of sorted points.
The dynamic program simply tries every possible location where the last cluster could begin and chooses
the split that minimizes the total within-cluster variance. This property does not hold in higher dimensions,
which is why the general k-means problem is much harder.

Algorithm:

1. Sort the n points in increasing order.

   x₁ ≤ x₂ ≤ ... ≤ xₙ

2. Precompute the clustering cost for every interval
   [i, j].

   Since every cluster consists of consecutive points,
   compute the cost of placing points
   xᵢ, xᵢ₊₁, ..., xⱼ
   into a single cluster.

   The optimal center is the mean of these points, so

   cost(i,j) =
   Σₜ₌ᵢʲ (xₜ − μᵢⱼ)²

   where

   μᵢⱼ = (xᵢ + ... + xⱼ)/(j−i+1)

3. Let DP[i][m] be the minimum clustering cost for the
   first i sorted points using exactly m clusters.

4. The recurrence is

   DP[i][m] =
   minₜ<m≤i
   { DP[t][m−1] + cost(t+1,i) }

   This means:

   - The first t points are optimally partitioned into
     m−1 clusters.

   - The last cluster contains points
     xₜ₊₁ through xᵢ.

5. Fill the DP table for all values of i and m.

6. The answer is DP[n][k].
"""
````

````python
def weighted_majority(experts, predictions, outcomes, gamma):
    n = len(experts)
    T = len(outcomes)
    # Initialize every expert with weight 1
    weights = [1 for _ in range(n)]
    final_predictions = []

    for t in range(T):
        # Experts make predictions for round t
        # predictions[t][i] = prediction of expert i
        q = predictions[t]

        # Calculate weighted votes
        upweight = 0
        downweight = 0

        for i in range(n):
            if q[i] == 1:
                upweight += weights[i]
            else:
                downweight += weights[i]

        # Majority prediction
        if upweight >= downweight:
            prediction = 1
        else:
            prediction = 0

        final_predictions.append(prediction)

        # Actual outcome is revealed
        outcome = outcomes[t]

        # Update weights
        for i in range(n):
            if q[i] != outcome:
                weights[i] *= (1 - gamma)

    return final_predictions
````

````python
"""
gradient of x^2 (derivative)
"""
def f_grad(x):
    return 2 * x
````

````python
"""
eta: learning rate
"""
def gradient_descent(f_grad, x0, eta, T):
    x = [0] * T
    x[0] = x0
    total = x0

    for t in range(T - 1):
        x[t + 1] = x[t] - eta * f_grad(x[t])
        total += x[t + 1]

    x_avg = total / T

    return x, x_avg
````

````python
def project(x):
    if x < 0:
        return 0
    elif x > 2:
        return 2
    else:
        return x
````

````python
"""
Ordinary gradient descent only asks:
"Where is the function decreasing?"

Projected gradient descent asks:
"Where is the function decreasing while staying inside the allowed region?"
"""
def gradient_descent_constrained(f_grad, x0, eta, T, project):
    x = [0] * T
    x[0] = x0
    total = x0

    for t in range(T - 1):
        # Unconstrained gradient step
        x_temp = x[t] - eta * f_grad(x[t])

        # Project back into the feasible set K
        x[t + 1] = project(x_temp)
        total += x[t + 1]

    x_avg = total / T

    return x, x_avg
````

# Newton's Method vs. Gradient Descent

Newton's method is an iterative algorithm used to find the **root** of a function, that is, a value $x^*$ such that

$$
f(x^*) = 0.
$$

Unlike gradient descent, which minimizes a function by moving in the direction of the negative gradient, Newton's method approximates the function locally with its tangent line and repeatedly jumps to where that tangent line crosses the x-axis.

---

## (a) Deriving Newton's Update Rule

Suppose we are currently at the point $x^{(t)}$.

The equation of the tangent line to $f(x)$ at this point is

$$
y = f'(x^{(t)})(x-x^{(t)}) + f(x^{(t)}).
$$

Newton's method chooses the x-intercept of this tangent line as the next approximation. Since the x-intercept occurs when $y=0$, we set

$$
0 = f'(x^{(t)})(x-x^{(t)}) + f(x^{(t)}).
$$

Solving for $x$ gives

$$
x-x^{(t)}
=
-\frac{f(x^{(t)})}{f'(x^{(t)})},
$$

or equivalently,

$$
\boxed{
x^{(t+1)}
=
x^{(t)}
-
\frac{f(x^{(t)})}{f'(x^{(t)})}
}
$$

This is the Newton update rule.

The intuition is simple: rather than taking a small step downhill like gradient descent, Newton's method asks, "If the tangent line were the entire function, where would its root be?" It then jumps directly to that point.

---

## (b) Error Analysis Using Taylor Expansion

Suppose the function has exactly one root $x^*$, and define the error at iteration $t$ as

$$
\epsilon^{(t)}
=
|x^{(t)}-x^*|.
$$

Since

$$
f(x^*)=0,
$$

we apply the Taylor expansion of $f(x^*)$ around the current approximation $x^{(t)}$:

$$
0
=
f(x^{(t)})
+
f'(x^{(t)})(x^*-x^{(t)})
+
\frac12
f''(y^{(t)})
(x^*-x^{(t)})^2,
$$

where $y^{(t)}$ lies between $x^{(t)}$ and $x^*$.

Substituting this expression into Newton's update rule and simplifying yields

$$
x^{(t+1)}-x^*
=
\frac{f''(y^{(t)})}
{2f'(x^{(t)})}
(x^*-x^{(t)})^2.
$$

Taking absolute values gives

$$
\boxed{
\epsilon^{(t+1)}
=
\frac{|f''(y^{(t)})|}
{2|f'(y^{(t)})|}
(\epsilon^{(t)})^2.
}
$$

This equation is extremely important because it describes how the error changes from one iteration to the next.

Notice that the new error depends on the **square** of the previous error.

---

## (c) Why Newton's Method is Quadratically Convergent

Suppose

$$
\frac{|f''(y^{(t)})|}
{2|f'(y^{(t)})|}
\le c
$$

for some constant $c$, and suppose the initial error satisfies

$$
\epsilon^{(0)}<1.
$$

Then

$$
\epsilon^{(t+1)}
\le
c(\epsilon^{(t)})^2.
$$

Expanding the first few iterations,

$$
\epsilon^{(1)}
\le
c\epsilon_0^2,
$$

$$
\epsilon^{(2)}
\le
c^3\epsilon_0^4,
$$

$$
\epsilon^{(3)}
\le
c^7\epsilon_0^8.
$$

The exponent on the error doubles every iteration.

In general,

$$
\epsilon^{(t)}
\le
c^{2^t-1}
(\epsilon^{(0)})^{2^t}.
$$

To achieve an error smaller than a desired tolerance $\Delta$, we require

$$
\epsilon^{(t)}
\le
\Delta.
$$

Solving for the number of iterations shows

$$
\boxed{
t
=
O\left(\log\log\frac1\Delta\right).
}
$$

This is called **quadratic convergence**.

The name comes from the fact that the error is squared at every iteration.

For example, if the error is

- $10^{-2}$,
- after one iteration it becomes approximately $10^{-4}$,
- then $10^{-8}$,
- then $10^{-16}$.

Each iteration roughly doubles the number of correct digits.

For comparison, gradient descent has linear convergence and typically requires

$$
O\left(\log\frac1\Delta\right)
$$

iterations, which is much slower.

---

## (d) Example: $f(x)=(x-3)^2$

Consider the function

$$
f(x)=(x-3)^2.
$$

This function has

- a minimum at $x=3$,
- and a root at $x=3$.

Assume the starting point is

$$
x^{(0)}=3.5.
$$

### Gradient Descent

The derivative is

$$
f'(x)=2(x-3).
$$

Using gradient descent,

$$
x^{(t+1)}
=
x^{(t)}
-
\eta f'(x^{(t)}).
$$

Choosing the common learning rate

$$
\eta=\frac12,
$$

gives

$$
x^{(t+1)}
=
x^{(t)}
-
(x^{(t)}-3)
=
3.
$$

Starting from

$$
x^{(0)}=3.5,
$$

we immediately obtain

$$
x^{(1)}=3.
$$

Therefore, gradient descent reaches the minimum in

$$
\boxed{1\text{ iteration}.}
$$

---

### Newton's Method

Newton's update is

$$
x^{(t+1)}
=
x^{(t)}
-
\frac{(x^{(t)}-3)^2}
{2(x^{(t)}-3)}
=
\frac{x^{(t)}+3}{2}.
$$

Starting from 3.5,

| Iteration | Approximation | Error |
|-----------|--------------|-------|
| 0 | 3.5000 | 0.5000 |
| 1 | 3.2500 | 0.2500 |
| 2 | 3.1250 | 0.1250 |
| 3 | 3.0625 | 0.0625 |
| 4 | 3.03125 | 0.03125 |

The error is cut in half every iteration.

Interestingly, Newton's method is **not quadratically convergent** for this function.

The reason is that quadratic convergence requires

$$
f'(x^*)\neq0.
$$

Here,

$$
f'(3)=0,
$$

so the root has multiplicity two (it is a repeated root).

Because the denominator of Newton's update becomes zero at the root, the assumptions behind the quadratic convergence proof no longer hold. Instead of squaring the error each iteration, Newton's method only halves it, resulting in **linear convergence**.

---

## Key Takeaways

- Gradient descent minimizes a function by following the negative gradient.
- Newton's method finds a root by repeatedly moving to the x-intercept of the tangent line.
- Newton's update rule is

$$
x^{(t+1)}
=
x^{(t)}
-
\frac{f(x^{(t)})}{f'(x^{(t)})}.
$$

- Under the assumptions that $f'(x^*)\neq0$ and $f''$ is continuous, Newton's method has **quadratic convergence**:

$$
\epsilon^{(t+1)}
=
O((\epsilon^{(t)})^2).
$$

- Therefore, Newton's method requires only

$$
O\left(\log\log\frac1\Delta\right)
$$

iterations to reach an accuracy of $\Delta$.

- However, if the root is repeated (as in $(x-3)^2$), then $f'(x^*)=0$, quadratic convergence is lost, and Newton's method converges only linearly.

- For the function $(x-3)^2$ starting at $x=3.5$, gradient descent with $\eta=\frac12$ reaches the minimum in one iteration, while Newton's method only cuts the error in half each iteration.

# Hedge Algorithm Mistake Bound

The Hedge algorithm is a randomized version of the Weighted Majority algorithm.

Unlike Weighted Majority, which always predicts according to the expert with the largest weight, Hedge chooses an expert randomly.

At round $t$, expert $i$ is selected with probability

$$
p_i^{(t)}
=
\frac{w_i^{(t)}}{Z^{(t)}},
$$

where

$$
Z^{(t)}
=
\sum_{i=1}^{n}w_i^{(t)}.
$$

If the selected expert makes a mistake, then Hedge also makes a mistake.

The update rule is

$$
w_i^{(t+1)}
=
w_i^{(t)}e^{-\epsilon},
$$

for every expert that made a mistake, where

$$
0<\epsilon<1.
$$

Correct experts keep their current weight.

---

## Step 1. Expected Mistake Probability

Let

$$
M^{(t)}
$$

be the set of experts that make a mistake during round $t$.

Since Hedge selects experts according to their normalized weights, the probability that Hedge makes a mistake is

$$
P_t
=
\sum_{i\in M^{(t)}}
\frac{w_i^{(t)}}{Z^{(t)}}.
$$

Therefore,

$$
P_t
$$

is exactly the expected number of mistakes made during round $t$.

The expected total number of mistakes over all $T$ rounds is

$$
\sum_{t=1}^{T}P_t.
$$

---

## Step 2. Analyze the Total Weight

Initially,

$$
w_i^{(1)}=1
$$

for every expert, so

$$
Z^{(1)}=n.
$$

After one update,

$$
Z^{(t+1)}
=
\sum_{i\notin M^{(t)}}w_i^{(t)}
+
e^{-\epsilon}
\sum_{i\in M^{(t)}}w_i^{(t)}.
$$

Factor out $Z^{(t)}$:

$$
Z^{(t+1)}
=
Z^{(t)}
\left(
1-(1-e^{-\epsilon})P_t
\right).
$$

Using the inequality

$$
1-x\le e^{-x},
$$

we obtain

$$
Z^{(t+1)}
\le
Z^{(t)}
e^{-(1-e^{-\epsilon})P_t}.
$$

Applying this repeatedly over all rounds,

$$
Z^{(T+1)}
\le
n
\exp
\left(
-(1-e^{-\epsilon})
\sum_{t=1}^{T}P_t
\right).
$$

---

## Step 3. Lower Bound Using the Best Expert

Suppose the best expert makes

$$
m^*
$$

mistakes.

Every mistake multiplies that expert's weight by

$$
e^{-\epsilon},
$$

so after $T$ rounds,

$$
w_{\text{best}}
=
e^{-\epsilon m^*}.
$$

Since the total weight must be at least the weight of the best expert,

$$
Z^{(T+1)}
\ge
e^{-\epsilon m^*}.
$$

---

## Step 4. Combine the Bounds

Since

$$
e^{-\epsilon m^*}
\le
n
\exp
\left(
-(1-e^{-\epsilon})
\sum P_t
\right),
$$

take logarithms:

$$
-\epsilon m^*
\le
\ln n
-
(1-e^{-\epsilon})
\sum P_t.
$$

Rearranging,

$$
\sum P_t
\le
\frac{\ln n+\epsilon m^*}
{1-e^{-\epsilon}}.
$$

---

## Step 5. Simplify the Denominator

Using the inequality

$$
e^{-\epsilon}
\le
1-\epsilon+\frac{\epsilon^2}{2},
$$

we obtain

$$
1-e^{-\epsilon}
\ge
\epsilon-\frac{\epsilon^2}{2}.
$$

Since

$$
\frac1{1-\epsilon/2}
\le
1+\epsilon,
$$

it follows that

$$
\frac1{1-e^{-\epsilon}}
\le
\frac1{\epsilon}+\!1.
$$

Substituting into the previous inequality gives

$$
\sum P_t
\le
m^*
+
\frac{\ln n}{\epsilon}
+
\epsilon T.
$$

Since

$$
\sum P_t
$$

is exactly the expected number of mistakes made by Hedge,

$$
\boxed{
\mathbb{E}[\text{Mistakes}]
\le
m^*
+
\frac{\ln n}{\epsilon}
+
\epsilon T.
}
$$

---

## Intuition

The proof follows the same "potential function" argument as the Weighted Majority algorithm.

The potential function is the total weight

$$
Z^{(t)}=\sum_i w_i^{(t)}.
$$

Every time Hedge is likely to make a mistake, a larger fraction of the total weight belongs to incorrect experts, causing the total weight to shrink.

On the other hand, the best expert loses weight only when it actually makes a mistake.

Comparing the upper bound on the total weight with the lower bound given by the best expert produces a bound on the expected number of mistakes.

Unlike Weighted Majority, Hedge is randomized, so the analysis bounds the **expected** number of mistakes rather than the exact number. The final guarantee shows that Hedge performs almost as well as the best expert, differing by only an additive term of

$$
\frac{\ln n}{\epsilon}+\epsilon T.
$$

# Lloyd's Procedure Does Not Always Find the Global Optimum (Even in One Dimension)

Lloyd's procedure (the standard k-means algorithm) is an iterative algorithm for solving the k-means clustering problem. At each iteration, it performs two steps:

1. **Assignment Step:** Assign every point to the nearest cluster center.
2. **Update Step:** Replace each cluster center with the mean of the points assigned to that cluster.

Although Lloyd's procedure is guaranteed to decrease (or leave unchanged) the objective function at every iteration and eventually terminate, it is **not guaranteed to find the global optimum**. Instead, it may converge to a **local optimum** that depends on the initial placement of the cluster centers.

---

## Counterexample

Consider the one-dimensional set of points

$$
S=\{1,2,4,5,9,10\}
$$

and suppose we wish to find

$$
k=2
$$

clusters.

Choose the initial centers to be

$$
4 \quad \text{and} \quad 5.
$$

Because both centers begin near the left side of the data, Lloyd's procedure converges to the clustering

$$
C_1=\{1,2,4\},
$$

$$
C_2=\{5,9,10\}.
$$

The corresponding cluster centers are

$$
\mu_1=\frac{1+2+4}{3}=\frac73\approx2.33,
$$

$$
\mu_2=\frac{5+9+10}{3}=8.
$$

At this point, reassigning the points does not change either cluster, so Lloyd's procedure terminates.

---

## Objective Function of Lloyd's Solution

The k-means objective function is

$$
f(S,C)
=
\sum_{x\in S}
(x-\mu(x))^2,
$$

where $\mu(x)$ is the center of the cluster containing point $x$.

For the first cluster,

$$
(1-2.33)^2+(2-2.33)^2+(4-2.33)^2
=
1.78+0.11+2.78
=
4.67.
$$

For the second cluster,

$$
(5-8)^2+(9-8)^2+(10-8)^2
=
9+1+4
=
14.
$$

Therefore,

$$
f(S,C)
=
4.67+14
=
18.67.
$$

---

## A Better Clustering

Now consider the clustering

$$
C_1=\{1,2,4,5\},
$$

$$
C_2=\{9,10\}.
$$

The centers are

$$
\mu_1=\frac{1+2+4+5}{4}=3,
$$

$$
\mu_2=\frac{9+10}{2}=9.5.
$$

Its objective function is

First cluster:

$$
(1-3)^2+(2-3)^2+(4-3)^2+(5-3)^2
=
4+1+1+4
=
10.
$$

Second cluster:

$$
(9-9.5)^2+(10-9.5)^2
=
0.25+0.25
=
0.5.
$$

Thus,

$$
f(S,C)
=
10.5.
$$

Since

$$
10.5<18.67,
$$

this clustering is strictly better than the one returned by Lloyd's procedure.

---

## Conclusion

Lloyd's procedure terminates because no point changes clusters after the final update. However, the clustering it returns does **not** minimize the k-means objective function.

Therefore,

$$
\boxed{\text{Lloyd's procedure can converge to a local optimum even in one dimension.}}
$$

This example demonstrates that the final clustering produced by Lloyd's algorithm depends on the initial placement of the cluster centers. Although the objective function decreases at every iteration, the algorithm is **not guaranteed to find the globally optimal clustering**.

````python
import random

def stochastic_gradient_descent(x, y, learning_rate=0.01, epochs=100):
    a = 0.0
    b = 0.0
    n = len(x)

    for epoch in range(epochs):
        # Pick one random point
        i = random.randint(0, n - 1)

        prediction = a * x[i] + b
        error = prediction - y[i]

        # SGD update
        a -= learning_rate * error * x[i]
        b -= learning_rate * error

    return a, b

# Example

x = [1,2,3,4,5]
y = [2,4,6,8,10]

a, b = stochastic_gradient_descent(x, y)

print("Slope:", a)
print("Intercept:", b)
````

# Stochastic Gradient Descent (SGD) for Linear Regression

Stochastic Gradient Descent (SGD) is a variant of gradient descent used when the training data is not available all at once or when the dataset is very large. Instead of computing the gradient using every training example, SGD updates the model parameters using **one randomly selected training example at a time**. This makes each update much faster and allows the algorithm to learn in an online setting where data arrives sequentially.

---

## Problem Setup

Suppose we want to fit the line

$$
f(x)=ax+b
$$

to a set of training examples

$$
S=\{(x_1,y_1),(x_2,y_2),\ldots,(x_T,y_T)\}.
$$

The goal is to choose the parameters \(a\) (the slope) and \(b\) (the intercept) that minimize the least-squares error.

The least-squares objective function is

$$
J(a,b)
=
\frac12
\sum_{i=1}^{T}
(ax_i+b-y_i)^2.
$$

In ordinary gradient descent, every iteration computes the gradient using **all** \(T\) training examples.

In stochastic gradient descent, however, one training example is chosen uniformly at random during each iteration, and only that example is used to update the parameters.

---

## Loss Function for One Training Example

Suppose the randomly selected training example is

$$
(x_i,y_i).
$$

The loss associated with this single example is

$$
\ell_i(a,b)
=
\frac12(ax_i+b-y_i)^2.
$$

Define the prediction error as

$$
e_i=ax_i+b-y_i.
$$

This quantity measures how far the current prediction is from the true value.

---

## Compute the Gradient

Differentiate the loss with respect to each parameter.

The derivative with respect to the slope is

$$
\frac{\partial \ell_i}{\partial a}
=
(ax_i+b-y_i)x_i.
$$

The derivative with respect to the intercept is

$$
\frac{\partial \ell_i}{\partial b}
=
ax_i+b-y_i.
$$

These derivatives form a stochastic estimate of the true gradient because they are computed from only one randomly selected training example.

---

## SGD Update Rule

Using a learning rate \(\eta\), the parameters are updated as

$$
a
\leftarrow
a-\eta(ax_i+b-y_i)x_i,
$$

$$
b
\leftarrow
b-\eta(ax_i+b-y_i).
$$

These updates are repeated many times, each time selecting a different random training example.

Unlike ordinary gradient descent, which performs one expensive update using the entire dataset, SGD performs many inexpensive updates using individual samples.

---

## Python Implementation

```python
import random

def stochastic_gradient_descent(x, y, learning_rate=0.01, epochs=1000):

    # Initialize parameters
    a = 0.0
    b = 0.0

    n = len(x)

    for epoch in range(epochs):

        # Randomly choose one training example
        i = random.randint(0, n - 1)

        prediction = a * x[i] + b
        error = prediction - y[i]

        # Update parameters
        a -= learning_rate * error * x[i]
        b -= learning_rate * error

    return a, b


# Example

x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

a, b = stochastic_gradient_descent(x, y)

print("Slope:", a)
print("Intercept:", b)
```

---

## Why This Works

Ordinary gradient descent computes the exact gradient

$$
\nabla J
=
\sum_{i=1}^{T}
\nabla \ell_i.
$$

Stochastic gradient descent computes the gradient using only one randomly selected example,

$$
\nabla \ell_i.
$$

Because each training example is selected uniformly at random,

$$
E[\nabla \ell_i]
=
\frac1T
\sum_{i=1}^{T}
\nabla \ell_i
=
\nabla J.
$$

Therefore, the stochastic gradient is an **unbiased estimate** of the true gradient.

Although each update is noisy, the average update points in the same direction as the true gradient.

---

## Expected Error Analysis

The analysis of SGD closely follows the proof for ordinary gradient descent.

For ordinary gradient descent, the optimization error after \(T\) iterations satisfies

$$
f(\bar{w})-f(w^*)
\le
\frac{LR^2}{2T},
$$

where

- \(L\) is an upper bound on the gradient,
- \(R\) is an upper bound on the distance from the optimum,
- \(T\) is the number of iterations.

Since the stochastic gradient is an unbiased estimator of the true gradient,

$$
E[g_t]
=
\nabla f(w_t),
$$

taking expectations throughout the gradient descent proof gives

$$
E[f(\bar{w})]-f(w^*)
\le
\frac{LR^2}{2T}.
$$

Therefore, the expected optimization error satisfies

$$
\boxed{
E[\text{Error}]
=
O\left(\frac{LR^2}{T}\right).
}
$$

---

## Intuition

Gradient descent computes the exact gradient using every training example before taking a step toward the optimum. This produces accurate updates but can be computationally expensive for large datasets.

Stochastic Gradient Descent instead computes the gradient using a single randomly selected training example. Each update is therefore much cheaper, although it introduces randomness into the optimization process.

The key insight is that the randomness averages out over many iterations. Since the stochastic gradient is an unbiased estimate of the true gradient, the algorithm still moves toward the optimum on average.

As the number of iterations increases, SGD converges to the optimal solution **in expectation**, and its expected optimization error decreases according to

$$
\boxed{
E[\text{Error}]
=
O\left(\frac{LR^2}{T}\right).
}
$$

This makes SGD one of the most widely used optimization algorithms in machine learning, especially for large datasets and online learning problems where training examples arrive one at a time.
